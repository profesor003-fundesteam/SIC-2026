## 1. Introducción: ¿Qué es un Diccionario?

En Python, un **diccionario** (`dict`) es la estructura de datos de mapeo (*mapping type*) nativa del lenguaje. A diferencia de las secuencias indexadas por posición numérica (como `tuple` o `list`), un diccionario almacena una colección de pares **clave: valor (`key: value`)**.

Sus características fundamentales son:

1. **Mapeo Clave-Valor:** Cada elemento consta de una **clave única** asociada a un **valor**.
2. **Claves Inmutables (*Hashables*):** Las claves deben ser objetos inmutables y *hashables* (`str`, `int`, `float`, `bool`, `tuple` de elementos inmutables). No pueden existir claves duplicadas.
3. **Valores Heterogéneos y Arbitrarios:** Los valores pueden ser de cualquier tipo de dato (inmutables o mutables, incluyendo listas, funciones u otros diccionarios) y sí pueden repetirse.
4. **Mutable y Dinámico:** Es un objeto **mutable**; permite insertar nuevos pares, modificar valores asociados a claves existentes y eliminar elementos *in-place* sin cambiar la identidad (`id()`) del diccionario.
5. **Ordenado por Inserción:** Desde Python 3.7 (como garantía oficial del lenguaje), los diccionarios **preservan el orden exacto en el que las claves fueron insertadas**.
6. **Acceso Ultrarrápido $O(1)$:** Gracias a su implementación mediante **tablas hash**, buscar, insertar o borrar una clave toma tiempo constante $O(1)$ en el caso promedio, sin importar si el diccionario tiene 10 o 10 millones de registros.

### Comparativa Rápida: Diccionarios (`dict`) vs. Listas (`list`) vs. Tuplas (`tuple`)

| Característica | Diccionario (`dict`) | Lista (`list`) | Tupla (`tuple`) |
| :--- | :--- | :--- | :--- |
| **Sintaxis literal** | `{"a": 1, "b": 2}` | `[1, 2, 3]` | `(1, 2, 3)` |
| **Tipo de acceso** | Por **clave** (`d["a"]`) | Por **índice entero** (`l[0]`) | Por **índice entero** (`t[0]`) |
| **Complejidad de búsqueda (`in`)** | **$O(1)$** promedio (Tabla Hash) | $O(n)$ (búsqueda lineal) | $O(n)$ (búsqueda lineal) |
| **Mutabilidad del contenedor** | **Mutable** | **Mutable** | **Inmutable** |
| **Orden de elementos** | Orden de inserción (Python 3.7+) | Orden posicional | Orden posicional |
| **¿Puede ser clave de otro `dict`?** | **No** (no es *hashable*) | **No** (no es *hashable*) | **Sí** (si sus elementos son *hashables*) |

## 2. Creación y Estructura de los Diccionarios

Python ofrece múltiples formas de instanciar un diccionario según la fuente de los datos. Es importante recordar que `{}` crea siempre un **diccionario vacío**, nunca un conjunto vacío (para un conjunto vacío se utiliza `set()`).

````python
# 1. Creación estándar mediante sintaxis literal {clave: valor}
usuario = {"nombre": "Ana", "edad": 28, "activo": True}
print("1. Literal con llaves:", usuario, "| Tipo:", type(usuario))

# 2. Diccionario vacío (¡Cuidado!: {} es un dict vacío, NO un set vacío)
dict_vacio_1 = {}
dict_vacio_2 = dict()
print("2. Diccionarios vacíos:", dict_vacio_1, dict_vacio_2, "| Longitud:", len(dict_vacio_1))

# 3. Uso del constructor dict() con argumentos por palabra clave (las claves deben ser identificadores válidos)
configuracion = dict(host="localhost", puerto=5432, ssl=True)
print("3. Constructor kwargs :", configuracion)

# 4. Uso del constructor dict() a partir de una secuencia de tuplas (clave, valor)
monedas = dict([("USD", "Dólar"), ("EUR", "Euro"), ("VES", "Bolívar")])
print("4. Desde pares (tuplas):", monedas)

# 5. Creación combinando dos secuencias paralelas con zip()
materias = ("Matemáticas", "Física", "Programación")
notas = [19, 18, 20]
boletin = dict(zip(materias, notas))
print("5. Con zip()          :", boletin)

# 6. Creación con dict.fromkeys(iterable, valor_por_defecto)
permisos_base = dict.fromkeys(["lectura", "escritura", "ejecucion"], True)
print("6. Con dict.fromkeys():", permisos_base)

# 7. Diccionarios heterogéneos y anidados
curso = {
    "codigo": "MAT-401",
    "titulo": "Redes Neuronales",
    "profesor": {"nombre": "Danny", "departamento": "Matemáticas"},
    "estudiantes_ids": (101, 102, 105)
}
print("\n--- Diccionario Anidado (Estructura estilo JSON) ---")
print(f"Curso: {curso['titulo']} | Profesor: {curso['profesor']['nombre']} | Inscritos: {len(curso['estudiantes_ids'])}")
````

### ¡Trampa Clásica con `dict.fromkeys()` y Valores Mutables!
Cuando usamos `dict.fromkeys(claves, valor)`, **todas las claves apuntan a la misma referencia en memoria de `valor`**. Si `valor` es una lista `[]`, modificar la lista de una clave afectará a todas las demás. Para evitarlo, se debe usar una comprensión de diccionario.

````python
# INCORRECTO: Todas las claves comparten el mismo objeto lista en memoria
grupos_mal = dict.fromkeys(["G1", "G2", "G3"], [])
grupos_mal["G1"].append("Pedro")
print("Problema con fromkeys y lista mutable:", grupos_mal)

# CORRECTO: Crear una lista nueva e independiente para cada clave
grupos_bien = {g: [] for g in ["G1", "G2", "G3"]}
grupos_bien["G1"].append("Pedro")
print("Solución con dict comprehension      :", grupos_bien)
````

## 3. Arquitectura Interna: ¿Por qué el Diccionario es Mutable pero sus Claves deben ser Inmutables?

### 3.1. El Diccionario como Objeto Mutable
Un diccionario (`dict`) es un **contenedor mutable**: después de ser creado en memoria, su estado interno puede cambiar libremente. Podemos agregar nuevas claves, cambiar los valores asociados o eliminar pares existentes **conservando la misma dirección de memoria (`id()`)**.

### 3.2. ¿Cómo funciona internamente un Diccionario en CPython? (Tablas Hash)
A diferencia de una lista que busca elementos recorriendo posiciones una por una ($O(n)$), un diccionario en CPython (`PyDictObject`) está implementado como una **Tabla Hash (*Hash Table*) con direccionamiento abierto (*Open Addressing*)**:

1. **Cálculo del Hash:** Cuando insertas o buscas `d[clave]`, Python invoca internamente `hash(clave)`, que devuelve un número entero de tamaño fijo (64 bits en arquitecturas modernas).
2. **Indexación Directa:** Con ese entero se aplica una máscara de bits (`indice = hash(clave) & mascara`) para saltar **directamente** a una ranura (*slot*) de un arreglo interno en memoria.
3. **Resolución de Colisiones:** Si dos claves distintas caen en la misma ranura (*colisión de hash*), Python utiliza una fórmula pseudoaleatoria de sondeo (*perturbation probing*) para encontrar la siguiente ranura libre.
4. **Tablas Compactas (Python 3.6+):** CPython divide el almacenamiento en dos arreglos:
   - `dk_indices`: Una tabla dispersa pequeña que solo guarda índices enteros.
   - `dk_entries`: Un arreglo denso que almacena las estructuras `(hash, clave_ptr, valor_ptr)` en el **orden exacto de inserción**. Esto redujo el consumo de memoria entre un 20% y un 25% y dio origen al orden garantizado de los diccionarios.

### 3.3. ¿Por qué las Claves DEBEN ser Inmutables (*Hashables*)?
Para que la tabla hash funcione, se deben cumplir dos leyes matemáticas invariantes durante toda la vida de la clave:
1. Si `a == b`, entonces obligatoriamente `hash(a) == hash(b)`.
2. El valor `hash(clave)` **nunca puede cambiar**.

Si Python permitiera usar un objeto **mutable** (como una `list`) como clave y luego modificaras esa lista con `.append()`, **su valor hash cambiaría**. Al intentar buscar la clave nuevamente en el diccionario, Python calcularía el nuevo hash, buscaría en una ranura distinta de la tabla y el objeto quedaría **perdido e inaccesible** dentro del diccionario. Por esta razón, **solo los objetos inmutables (*hashables*) pueden ser claves**.

````python
# 1. Demostración de la mutabilidad del diccionario (su ID en memoria no cambia)
inventario = {"manzanas": 10, "peras": 5}
id_inicial = id(inventario)
print("ID inicial del diccionario:", id_inicial)

inventario["naranjas"] = 20      # Inserción in-place
inventario["manzanas"] = 15      # Modificación in-place
print(f"ID tras agregar y mutar   : {id(inventario)} (¿Es el mismo objeto en memoria?: {id(inventario) == id_inicial})")

# 2. Claves válidas (Objetos inmutables y hashables)
diccionario_valido = {
    "texto": "String",
    42: "Entero",
    3.1416: "Flotante",
    (10, 20): "Tupla inmutable"
}
print("\n--- Claves Válidas (Inmutables / Hashables) ---")
print(diccionario_valido)

# 3. Claves inválidas (Objetos mutables o que contienen mutables)
print("\n--- Intentos con Claves Inválidas (Mutables / No Hashables) ---")
try:
    {[1, 2]: "Lista como clave"}
except TypeError as e:
    print(f"Usar una lista como clave falló -> {type(e).__name__}: {e}")

try:
    {{"a": 1}: "Diccionario como clave"}
except TypeError as e:
    print(f"Usar un dict como clave falló  -> {type(e).__name__}: {e}")

try:
    {(1, [2, 3]): "Tupla con lista mutable adentro"}
except TypeError as e:
    print(f"Usar tupla con lista interna   -> {type(e).__name__}: {e}")
````

### 3.4. Caso Especial: Colisión de Igualdad entre `1`, `True` y `1.0`
En Python, `1 == True == 1.0` evalúa a `True` y además comparten exactamente el mismo hash (`hash(1) == hash(True) == hash(1.0) == 1`).
- Para un diccionario, **`1`, `True` y `1.0` son exactamente la misma clave**.
- Si insertas los tres en un diccionario, **se conserva la primera clave insertada**, pero **su valor se sobrescribe** con las asignaciones posteriores.

````python
import timeit

# Demostración de claves equivalentes (1, True, 1.0)
print("¿1 == True == 1.0?          :", 1 == True == 1.0)
print("Hashes (1, True, 1.0)       :", hash(1), hash(True), hash(1.0))

colision_claves = {1: "Uno entero", True: "Booleano True", 1.0: "Uno flotante"}
print("Diccionario resultante      :", colision_claves)

# Demostración empírica de velocidad de búsqueda O(1) en dict vs O(n) en tuple
n = 100_000
datos_dict = {i: None for i in range(n)}
datos_tupla = tuple(range(n))
objetivo = n - 1

t_dict = timeit.timeit(lambda: objetivo in datos_dict, number=1000)
t_tupla = timeit.timeit(lambda: objetivo in datos_tupla, number=1000)

print("\n--- Comparación de Rendimiento de Búsqueda: O(1) vs O(n) ---")
print("Buscar 1,000 veces el último elemento (100,000 datos):")
print(f"  Tiempo en Diccionario O(1): {t_dict:.5f}s")
print(f"  Tiempo en Tupla       O(n): {t_tupla:.5f}s")
````

## 4. Acceso, Inserción, Modificación y Eliminación de Elementos

### 4.1. Acceso a Valores: Corchetes `d[clave]` vs. Método `.get()`
- **`d[clave]`**: Devuelve el valor asociado a `clave`. Si la clave **no existe**, lanza inmediatamente una excepción `KeyError`.
- **`d.get(clave, valor_por_defecto=None)`**: Devuelve el valor asociado a `clave` si existe; de lo contrario, retorna `valor_por_defecto` (`None` si no se especifica) **sin lanzar ningún error**.

````python
servidor = {"hostname": "srv-produccion", "ip": "192.168.1.10", "estado": "activo"}

# 1. Acceso directo con []
print("Acceso directo servidor['ip']:", servidor["ip"])

try:
    print(servidor["puerto"])
except KeyError as e:
    print(f"Error con corchetes en clave inexistente: {type(e).__name__} -> {e}")

# 2. Acceso seguro con .get()
print("Con .get('puerto') sin default          :", servidor.get("puerto"))
print("Con .get('puerto', 8080) con default    :", servidor.get("puerto", 8080))
````

### 4.2. Inserción, Actualización y Eliminación (`del`, `.pop()`, `.popitem()`, `.clear()`)

| Operación / Método | Comportamiento |
| :--- | :--- |
| `d[clave] = valor` | Si `clave` no existe, la crea al final. Si ya existe, sobrescribe su valor manteniendo su posición original. |
| `del d[clave]` | Elimina el par asociado a `clave`. No devuelve el valor. Lanza `KeyError` si no existe. |
| `d.pop(clave[, default])` | Elimina `clave` y **devuelve su valor**. Si no existe, devuelve `default` (o lanza `KeyError` si no se indicó `default`). |
| `d.popitem()` | Elimina y devuelve una tupla `(clave, valor)` con el **último par insertado** (orden LIFO desde Python 3.7+). Lanza `KeyError` si el diccionario está vacío. |
| `d.clear()` | Vacía por completo el diccionario *in-place*, dejándolo en `{}`. |

````python
cartera = {"A": 10, "B": 20, "C": 30}
print("Inicial                  :", cartera)

# 1. Inserción y actualización
cartera["D"] = 40   # Nueva clave
cartera["B"] = 25   # Sobrescribe valor existente
print("Tras insertar 'D' y mod 'B':", cartera)

# 2. Eliminación con sentencia del
del cartera["A"]
print("Tras del cartera['A']    :", cartera)

# 3. Extracción con .pop(clave, default)
valor_c = cartera.pop("C")
print(f"Valor extraído con pop('C'): {valor_c} | Cartera: {cartera}")
valor_inexistente = cartera.pop("Z", 0)  # Evita KeyError devolviendo 0
print("pop('Z', 0) seguro       :", valor_inexistente)

# 4. Extracción LIFO (último insertado) con .popitem()
ultimo_par = cartera.popitem()
print(f"Último par con popitem() : {ultimo_par} | Cartera: {cartera}")

# 5. Vaciado total con .clear()
cartera.clear()
print("Tras .clear()            :", cartera)
````

## 5. Métodos Nativos de los Diccionarios (`dict`)

A diferencia de las tuplas (que solo tienen 2 métodos), la clase `dict` dispone de **11 métodos públicos** diseñados para consultar vistas dinámicas, actualizar lotes de datos y gestionar valores por defecto:

`clear`, `copy`, `fromkeys`, `get`, `items`, `keys`, `pop`, `popitem`, `setdefault`, `update`, `values`.

### 5.1. Objetos de Vista Dinámica (`keys()`, `values()`, `items()`)
Los métodos `.keys()`, `.values()` e `.items()` **no devuelven listas estáticas**, sino objetos de vista (`dict_keys`, `dict_values`, `dict_items`):
- **Son dinámicos y ligeros:** No duplican los datos en memoria; si el diccionario cambia, la vista refleja el cambio automáticamente.
- **Operaciones de Conjuntos (`set`):** Tanto `d.keys()` como `d.items()` (si los valores son *hashables*) se comportan como conjuntos matemáticos y soportan intersección (`&`), unión (`|`), diferencia (`-`) y diferencia simétrica (`^`).

````python
# Lista completa de métodos públicos de dict
metodos_dict = [m for m in dir(dict) if not m.startswith("_")]
print("Métodos públicos de 'dict':", metodos_dict)

hardware = {"CPU": "Ryzen 7", "RAM": "32GB"}
vista_claves = hardware.keys()  # Guardamos la vista antes de modificar el diccionario

print("\n--- Vistas Dinámicas en Tiempo Real ---")
print("Vista antes de insertar  :", vista_claves)

hardware["GPU"] = "RTX 4070"
print("Vista después de insertar:", vista_claves)
print("Valores (.values())      :", hardware.values())
print("Pares (.items())         :", hardware.items())

# Operaciones algebraicas de conjuntos sobre .keys()
d1 = {"a": 1, "b": 2, "c": 3}
d2 = {"b": 20, "c": 3, "d": 4}

print("\n--- Operaciones de Conjuntos sobre .keys() ---")
print("Claves en común (&)      :", d1.keys() & d2.keys())
print("Claves solo en d1 (-)    :", d1.keys() - d2.keys())
````

### 5.2. Uso de `.setdefault()`, `.update()` y Copias (`.copy()` vs `deepcopy`)

- **`d.setdefault(clave, default=None)`**:
  - Si `clave` **ya está** en el diccionario, devuelve su valor actual **sin modificar nada**.
  - Si `clave` **no está**, la inserta en el diccionario con el valor `default` y devuelve ese valor. Es sumamente útil para inicializar listas al agrupar datos.
- **`d.update(otro)`**: Actualiza el diccionario *in-place* con los pares de otro diccionario, de un iterable de pares `(clave, valor)` o de argumentos por palabra clave.
- **`d.copy()` vs `copy.deepcopy(d)`**: `.copy()` realiza una **copia superficial (*shallow copy*)**; si el diccionario contiene listas o diccionarios anidados, las referencias internas se comparten. Para clonar recursivamente toda la estructura se requiere `copy.deepcopy()`.

````python
import copy

# 1. Patrón de agrupación usando .setdefault()
alimentos = [("Fruta", "Manzana"), ("Verdura", "Zanahoria"), ("Fruta", "Cambur"), ("Verdura", "Espinaca")]
categorias = {}

for tipo, nombre in alimentos:
    # Si 'tipo' no existe, crea una lista vacía [] y luego hace .append(nombre)
    categorias.setdefault(tipo, []).append(nombre)

print("Agrupación con .setdefault():", categorias)

# 2. Actualización masiva con .update()
preferencias = {"tema": "claro", "idioma": "es"}
preferencias.update({"tema": "oscuro", "fuente": 14}, autoguardado=True)
print("Tras .update()              :", preferencias)

# 3. Diferencia entre .copy() (superficial) y copy.deepcopy() (profunda)
original = {"usuario": "Luis", "roles": ["editor"]}
copia_superficial = original.copy()
copia_profunda = copy.deepcopy(original)

copia_superficial["roles"].append("admin")       # Modifica la lista compartida
copia_profunda["roles"].append("superusuario")   # Modifica solo la lista clonada

print("\n--- Copia Superficial (.copy) vs Copia Profunda (deepcopy) ---")
print("Original tras mutar copia_superficial:", original, "(¡Se alteró el original!)")
print("Original tras mutar copia_profunda   :", original, "(Intacto)")
````

## 6. Operadores de Fusión (`|`, `|=`) y Desempaquetado (`**`)

Desde **Python 3.9 (PEP 584)**, los diccionarios incorporan operadores nativos de unión:

- **Fusión (`d1 | d2`)**: Crea y devuelve un **nuevo diccionario** combinando ambos. Si hay claves repetidas, **prevalece el valor del diccionario de la derecha (`d2`)**.
- **Actualización *in-place* (`d1 |= d2`)**: Equivalente a `d1.update(d2)`, modifica `d1` directamente.
- **Desempaquetado con doble asterisco (`**`)**: Permite expandir los pares clave-valor dentro de otro literal de diccionario `{**d1, **d2}` o pasarlos como argumentos con nombre (`**kwargs`) a una función.

````python
d_base = {"timeout": 30, "reintentos": 3, "debug": False}
d_custom = {"reintentos": 5, "debug": True, "log": "info"}

# 1. Operador de unión | (Crea un nuevo diccionario; prevalece la derecha en colisiones)
config_final = d_base | d_custom
print("Fusión con | (d_base | d_custom) :", config_final)

# 2. Desempaquetado de diccionarios con **
config_extendida = {**d_base, **d_custom, "version": 2}
print("Fusión con desempaquetado **     :", config_extendida)

# 3. Uso de ** para pasar un diccionario como argumentos de una función
def conectar_bd(servidor, puerto, usuario):
    print(f"Conectando a {servidor} en el puerto {puerto} (Usuario: {usuario})")

credenciales = {"servidor": "db_local", "puerto": 5432, "usuario": "admin"}
conectar_bd(**credenciales)

# 4. Operadores de pertenencia (in / not in) -> Evalúan SIEMPRE sobre las CLAVES en O(1)
print("\n--- Operadores de Pertenencia (in / not in) ---")
print("¿Está la clave 'timeout' en d_base? :", "timeout" in d_base)
print("¿Está el valor 30 en d_base?        :", 30 in d_base, "(¡'in' busca solo en las claves!)")
print("¿Está el valor 30 en d_base.values()?:", 30 in d_base.values())
````

## 7. Iteración y Funciones Integradas (*Built-in Functions*) Asociadas

Cuando pasamos un diccionario directamente a un bucle `for x in d:` o a funciones como `min(d)`, `max(d)`, `sorted(d)`, Python **opera por defecto sobre sus claves**.

| Función / Patrón | Comportamiento sobre un Diccionario `d` |
| :--- | :--- |
| `len(d)` | Retorna el número de pares clave-valor almacenados ($O(1)$). |
| `for k, v in d.items():` | Recorre simultáneamente cada clave y su valor desempaquetando la tupla `(k, v)`. |
| `max(d, key=d.get)` | Retorna la **clave** que posee el valor máximo dentro del diccionario. |
| `sorted(d.items(), key=lambda x: x[1])` | Devuelve una lista de tuplas `(clave, valor)` ordenada según sus valores. |
| `reversed(d)` | Itera sobre las claves en orden inverso de inserción (Python 3.8+). |
| `sum(d.values())` | Suma todos los valores numéricos del diccionario. |

````python
ventas = {
    "Sofía": 14200,
    "Mateo": 9800,
    "Valentina": 18500,
    "Diego": 12100
}

# 1. Iteración pitónica con .items()
print("--- Iteración desempaquetando .items() ---")
for vendedor, monto in ventas.items():
    print(f"  Vendedor: {vendedor:<8} | Ventas: ${monto:,}")

# 2. Funciones integradas y búsqueda por criterio (key=ventas.get)
mejor_vendedor = max(ventas, key=ventas.get)
peor_vendedor = min(ventas, key=ventas.get)

# Ordenar un diccionario de mayor a menor según sus valores
ventas_ordenadas = dict(sorted(ventas.items(), key=lambda par: par[1], reverse=True))

print("\n--- Funciones Integradas ---")
print("Total de vendedores len()           :", len(ventas))
print(f"Suma total de ventas sum(.values()) : ${sum(ventas.values()):,}")
print(f"Vendedor con mayor venta (key=d.get): {mejor_vendedor} (${ventas[mejor_vendedor]:,})")
print(f"Vendedor con menor venta (key=d.get): {peor_vendedor} (${ventas[peor_vendedor]:,})")
print("Diccionario ordenado por ventas DESC:", ventas_ordenadas)
````

## 8. Comprensión de Diccionarios (*Dict Comprehensions*)

A diferencia de las tuplas, los diccionarios **sí tienen su propia sintaxis de comprensión nativa** utilizando llaves `{}` y separando la expresión de la clave y la del valor mediante dos puntos `:`:

```python
{expresion_clave: expresion_valor for elemento in iterable if condicion}
```

````python
# 1. Construcción matemática: número -> cubo
cubos = {n: n**3 for n in range(1, 6)}
print("1. Cubos del 1 al 5        :", cubos)

# 2. Transformación de valores sobre un diccionario existente
precios_usd = {"Monitor": 200, "Teclado": 45, "Cable HDMI": 8, "Silla": 150}
precios_con_iva = {prod: round(precio * 1.16, 2) for prod, precio in precios_usd.items()}
print("2. Precios con IVA (16%)   :", precios_con_iva)

# 3. Filtrado condicional de pares
productos_costosos = {k: v for k, v in precios_usd.items() if v >= 50}
print("3. Filtrado (>= $50)       :", productos_costosos)

# 4. Inversión de claves y valores (intercambiar llave por valor)
codigos_pais = {"Venezuela": "VE", "Colombia": "CO", "Argentina": "AR"}
pais_por_codigo = {codigo: pais for pais, codigo in codigos_pais.items()}
print("4. Diccionario invertido   :", pais_por_codigo)
````

## 9. Nivel Avanzado: Diccionarios Especializados e Inmutables (`collections` y `types`)

La biblioteca estándar de Python amplía el poder de `dict` con estructuras especializadas:

1. **`collections.defaultdict(fabrica)`**: Subclase de `dict` que invoca automáticamente a `fabrica()` (por ejemplo `int`, `list`, `set`) para crear un valor por defecto cada vez que se accede a una clave que aún no existe.
2. **`collections.Counter(iterable)`**: Subclase de `dict` diseñada para contar frecuencias de elementos hashables.
3. **`types.MappingProxyType(diccionario)`**: ¿Qué pasa si necesitamos pasar un diccionario a una función pero **queremos que sea inmutable (solo lectura)**? `MappingProxyType` crea un envoltorio dinámico de solo lectura que impide cualquier modificación a través del proxy.

````python
from collections import defaultdict, Counter
from types import MappingProxyType

# 1. Uso de defaultdict(list) para agrupar sin verificar si la clave existe
usuarios_por_rol = defaultdict(list)
registros = [("admin", "root"), ("invitado", "guest1"), ("admin", "danny")]

for rol, user in registros:
    usuarios_por_rol[rol].append(user)  # No lanza KeyError jamás

print("1. defaultdict(list)  :", dict(usuarios_por_rol))

# 2. Uso de Counter para análisis de frecuencias
frecuencias = Counter("abracadabra")
print("2. Counter frecuencias:", frecuencias)
print("   Top 2 más comunes  :", frecuencias.most_common(2))

# 3. Diccionario de solo lectura (Inmutable) con MappingProxyType
config_interna = {"MODO": "produccion", "TIMEOUT": 10}
config_inmutable = MappingProxyType(config_interna)

print("3. MappingProxyType   :", config_inmutable)
try:
    config_inmutable["TIMEOUT"] = 99
except TypeError as e:
    print(f"   Intento de mutar MappingProxyType -> {type(e).__name__}: {e}")
````

## 10. Resumen y Buenas Prácticas: ¿Cuándo usar Diccionarios?

1. **Búsquedas instantáneas por identificador ($O(1)$):** Si necesitas consultar información repetidamente mediante un código, ID, cédula o nombre de usuario, un diccionario es órdenes de magnitud más rápido que recorrer una lista o tupla.
2. **Datos estructurados y semánticos (estilo JSON):** Ideales para representar entidades con atributos con nombre, configuraciones, respuestas de APIs REST y parámetros dinámicos (`**kwargs`).
3. **Conteo, indexación y agrupación:** Combinados con `defaultdict` o `Counter`, permiten construir histogramas, índices invertidos y cachés en pocas líneas de código.
4. **Sinergia entre Tuplas y Diccionarios:** Recuerda que las **tuplas** y los **diccionarios** se complementan perfectamente: cuando necesites indexar un diccionario por una clave compuesta (como `(latitud, longitud)` o `(año, mes)`), utiliza una **tupla inmutable como clave del diccionario**.
