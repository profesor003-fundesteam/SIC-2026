## 1. Apartado Inicial: Los Decoradores Explicados de Forma Súper Sencilla (`@`)

### ¿Qué es un decorador en palabras simples?
Imagina que tienes un **teléfono celular** (esa es tu **función original**). El teléfono ya cumple su tarea: hacer llamadas.
Ahora imagina que le pones una **funda protectora** (ese es el **decorador**):
* El teléfono sigue siendo el mismo por dentro y no tuviste que desarmarlo.
* Pero ahora, gracias a la funda, tiene una capacidad extra (protección contra caídas).

En Python, **un decorador es simplemente una función que "envuelve" a otra función para ejecutar código extra antes y/o después de ella, sin tener que modificar el código de la función original.**

### ¿Cómo se lee y cómo funciona?
1. Creas una función "envoltorio" que recibe una `funcion` como entrada.
2. Dentro de ella defines qué quieres que pase **antes**, llamas a la `funcion()` original, y defines qué quieres que pase **después**.
3. Para aplicarlo sobre cualquier función, solo colocas `@nombre_del_decorador` justo arriba de `def`.

> **¿Por qué aprender decoradores antes de ver Clases?**
> Porque dentro de las clases en Python usarás decoradores todo el tiempo (`@classmethod`, `@staticmethod`, `@property`, `@dataclass`). ¡Entender qué hace el símbolo `@` te quitará cualquier duda!

```python
# Ejemplo 1.1: El decorador más sencillo y fácil de entender

# 1. Creamos el decorador (la "funda" o envoltorio)
def poner_marco_bonito(funcion_original):
    def envoltura():
        print("========================================")  # Se ejecuta ANTES
        funcion_original()                                  # Se ejecuta tu función original
        print("========================================\n") # Se ejecuta DESPUÉS
    return envoltura


# 2. Usamos @poner_marco_bonito encima de cualquier función para "decorarla"
@poner_marco_bonito
def mostrar_bienvenida():
    print("  ¡Bienvenido al sistema de inventario!")

@poner_marco_bonito
def mostrar_despedida():
    print("  Sesión finalizada correctamente.")


# 3. Al llamar a las funciones, automáticamente salen con su marco:
mostrar_bienvenida()
mostrar_despedida()
```

### ¿Y si la función que quiero decorar recibe datos (parámetros) o devuelve un resultado (`return`)?
Para que tu decorador funcione con **cualquier función** (sin importar cuántos datos reciba), en la función interna `envoltura` usamos `*args, **kwargs` (que significa *"recibe todos los argumentos que vengan y pásaselos tal cual a la función original"*).

```python
# Ejemplo 1.2: Decorador universal que funciona con cualquier función con parámetros y return

def avisar_ejecucion(funcion_original):
    def envoltura(*args, **kwargs):
        print(f"-> [AVISO] Ejecutando '{funcion_original.__name__}' con datos: {args}")
        resultado = funcion_original(*args, **kwargs)  # Guardamos el resultado original
        print(f"-> [LISTO] '{funcion_original.__name__}' devolvió: {resultado}\n")
        return resultado  # Entregamos el resultado para que no se pierda
    return envoltura


@avisar_ejecucion
def sumar(a, b):
    return a + b

@avisar_ejecucion
def calcular_descuento(precio, porcentaje):
    return precio - (precio * porcentaje / 100)


total_suma = sumar(15, 25)
precio_oferta = calcular_descuento(200, 15)
```

## 2. Resumen Breve: ¿Para Qué Sirven las Funciones y Buenas Prácticas?

Antes de dar el salto a las clases, repasemos de forma muy concreta el propósito de las funciones y cómo escribirlas bien:

### ¿Para qué sirven las funciones?
* **Evitar repetir código (Principio DRY)**: Escribes una lógica una sola vez y la reutilizas en todo tu programa.
* **Dividir y vencer**: Transforman un problema grande y confuso en pequeñas piezas fáciles de entender y mantener.
* **Facilitar las pruebas y corrección de errores**: Si algo falla en el cálculo de impuestos, solo revisas la función `calcular_impuesto()`, no todo el programa.
* **Hacer el código legible**: Llamar a `validar_usuario(datos)` explica por sí solo qué está haciendo el programa en esa línea.

### Checklist Rápido de Buenas Prácticas en Funciones
1. **Una sola tarea por función**: Si tu función calcula, guarda en base de datos y envía un correo, divídela en 3 funciones pequeñas.
2. **Nombres claros con verbos (`snake_case`)**: Usa `calcular_total()`, `obtener_cliente()` o `es_valido()` en lugar de nombres vagos como `proceso1()` o `hacer_algo()`.
3. **Devuelve valores con `return`, no con `print()`**: Usa `return` para que el resultado pueda ser aprovechado por el resto del programa.
4. **Cero listas vacías como valor por defecto**: Usa `def agregar(item, lista=None):` en vez de `lista=[]` para no compartir la misma lista por error entre llamadas.
5. **Agrega *Type Hints* y un *Docstring* corto**: Indicar qué tipos de datos entran y qué tipo sale (`def area(base: float, altura: float) -> float:`) hace que tu código sea autodocumentado.

```python
# Ejemplo 2: Modelo rápido de una función que aplica todas las buenas prácticas

def calcular_cuota_mensual(monto_prestamo: float, meses: int, tasa_mensual: float = 0.02) -> float:
    """Calcula la cuota mensual fija de un préstamo con interés simple mensual."""
    if monto_prestamo <= 0 or meses <= 0:
        raise ValueError("El monto y los meses deben ser mayores a cero.")

    interes_total = monto_prestamo * tasa_mensual * meses
    return round((monto_prestamo + interes_total) / meses, 2)


print(f"Cuota mensual por préstamo de $1,200 a 12 meses: ${calcular_cuota_mensual(1200, 12)}")
```

## 3. Fundamentos de las Clases en Python (Programación Orientada a Objetos)

### 3.1 ¿Qué es una Clase y qué es un Objeto (Instancia)?
Cuando un programa crece, tener datos sueltos por un lado (diccionarios, listas) y funciones sueltas por otro se vuelve caótico. Las **Clases** resuelven esto uniendo **Datos (Atributos)** + **Comportamientos (Métodos)** en una sola estructura:

* **Clase (`class`)**: Es el **molde, plano o plantilla**. Define qué características (atributos) y qué acciones (métodos) tendrán los objetos de ese tipo.
  * *Convención de nombres*: Las clases siempre se escriben en **`PascalCase`** (primera letra de cada palabra en mayúscula, sin guiones bajos: `CuentaBancaria`, `ClienteVip`, `CarritoCompras`).
* **Objeto o Instancia**: Es cada **elemento concreto** creado a partir de ese molde. Con una sola clase `Automovil` puedes crear 100 objetos `Automovil` distintos (cada uno con su propio color, marca y velocidad).

### 3.2 ¿Qué significa `self`?
Es la duda número uno al aprender clases en Python, y su explicación es muy simple:
* **`self` significa "este mismo objeto"**.
* Si creas dos perros con la clase `Perro`: `perro1 = Perro("Rex")` y `perro2 = Perro("Luna")`, cuando haces `perro1.ladrar()`, internamente Python pasa a `perro1` como primer argumento (`self`). Así el método sabe si debe imprimir `"Rex está ladrando"` o `"Luna está ladrando"`.

### 3.3 Atributos de Instancia vs. Atributos de Clase
* **Atributos de Instancia (`self.atributo`)**: Se definen dentro de `__init__`. Cada objeto tiene su propia copia independiente (ej. el `nombre` o `saldo` de cada cliente).
* **Atributos de Clase**: Se definen directamente dentro de la clase, fuera de cualquier método. Su valor es **compartido por absolutamente todos los objetos** de esa clase (ej. el nombre del banco o la tasa de IVA general).

```python
# Ejemplo 3: Creación de una Clase, Atributos de Clase vs. Instancia y qué hace 'self'

class CuentaBancaria:
    # ATRIBUTO DE CLASE (compartido por todas las cuentas)
    banco = "Banco Central Python"
    comision_retiro = 1.50

    # MÉTODO INICIALIZADOR (__init__): Se ejecuta automáticamente al crear un objeto nuevo
    def __init__(self, titular: str, saldo_inicial: float = 0.0):
        # ATRIBUTOS DE INSTANCIA (propios de cada cuenta individual usando 'self')
        self.titular = titular
        self.saldo = saldo_inicial

    # MÉTODO DE INSTANCIA: Acción que puede realizar cada cuenta
    def mostrar_resumen(self):
        print(f"[{self.banco}] Titular: {self.titular:<15} | Saldo disponible: ${self.saldo:,.2f}")


# Creamos 2 OBJETOS (instancias) distintos usando el mismo molde (Clase CuentaBancaria):
cuenta_ana = CuentaBancaria("Ana Morales")
cuenta_luis = CuentaBancaria("Luis Pineda", 320.0)

# Cada objeto tiene sus propios datos en 'self', pero comparten el atributo de clase 'banco':
cuenta_ana.mostrar_resumen()
cuenta_luis.mostrar_resumen()
```

## 4. Énfasis Especial: Todos los Métodos que Puedes Programar en tus Clases

Cuando diseñas una clase en Python, no todos los métodos hacen lo mismo. Tienes a tu disposición **5 grandes herramientas** para programar el comportamiento de tus clases:

1. **Métodos de Instancia (`self`)**: Acciones normales que trabajan con los datos de un objeto individual.
2. **Métodos de Clase (`@classmethod` y `cls`)**: Métodos que trabajan con la clase como un todo o sirven como **fábricas de objetos (constructores alternativos)**.
3. **Métodos Estáticos (`@staticmethod`)**: Funciones auxiliares o validadores que viven dentro de la clase por orden, pero no usan ni `self` ni `cls`.
4. **Propiedades (`@property` y `@atributo.setter`)**: Métodos disfrazados de atributos para calcular valores al vuelo y proteger datos con validaciones.
5. **Métodos Mágicos o Especiales (*Dunder Methods* `__metodo__`)**: Métodos que conectan tus objetos con los símbolos y funciones nativas de Python (`+`, `==`, `len()`, `print()`, `[]`, etc.).

Veamos cada uno en detalle con ejemplos claros.

---

### 4.1 Los 3 Tipos Clásicos: Instancia (`self`), de Clase (`@classmethod`) y Estáticos (`@staticmethod`)

| Tipo de Método | Decorador | Primer Parámetro | ¿A qué tiene acceso? | ¿Para qué se programa principalmente? |
| :--- | :--- | :--- | :--- | :--- |
| **De Instancia** | *(Ninguno)* | `self` (el objeto) | Atributos del objeto (`self.x`) y de la clase | Leer o modificar el estado de un objeto específico (depositar, vender, actualizar). |
| **De Clase** | `@classmethod` | `cls` (la clase) | Solo a la clase (`cls`), no a un objeto individual | Crear **constructores alternativos** (ej. crear un objeto desde un texto, un JSON o una fecha) o cambiar configuraciones globales de la clase. |
| **Estático** | `@staticmethod` | *(Ninguno)* | Ni a `self` ni a `cls` | Funciones utilitarias o de validación relacionadas con el tema de la clase (ej. validar formato de correo o convertir monedas). |

```python
# Ejemplo 4.1: Comparativa práctica de Métodos de Instancia, @classmethod y @staticmethod

class Empleado:
    # Atributo de clase: porcentaje de aumento anual para toda la empresa
    aumento_anual = 1.08  # 8% de aumento

    def __init__(self, nombre: str, cargo: str, salario_mensual: float):
        self.nombre = nombre
        self.cargo = cargo
        self.salario_mensual = salario_mensual

    # --------------------------------------------------------------------------
    # 1. MÉTODOS DE INSTANCIA (Usan 'self': afectan solo a ESTE empleado)
    # --------------------------------------------------------------------------
    def aplicar_aumento_individual(self):
        """Modifica el salario de este empleado usando el porcentaje de la clase."""
        self.salario_mensual = round(self.salario_mensual * self.aumento_anual, 2)

    def obtener_ficha(self) -> str:
        return f"Empleado: {self.nombre:<14} | Cargo: {self.cargo:<15} | Salario: ${self.salario_mensual:,.2f}"

    # --------------------------------------------------------------------------
    # 2. MÉTODOS DE CLASE (Usan '@classmethod' y 'cls': operan sobre la Clase)
    # --------------------------------------------------------------------------
    @classmethod
    def fijar_nuevo_aumento_anual(cls, nuevo_factor: float):
        """Cambia el porcentaje de aumento anual para TODOS los empleados de la clase."""
        cls.aumento_anual = nuevo_factor

    @classmethod
    def desde_texto_guiones(cls, texto: str):
        """
        CONSTRUCTOR ALTERNATIVO:
        Permite crear un objeto Empleado directamente desde una cadena 'Nombre-Cargo-Salario'.
        Usa cls(...) que equivale a llamar a Empleado(...).
        """
        nombre, cargo, salario_str = texto.split("-")
        return cls(nombre.strip(), cargo.strip(), float(salario_str))

    # --------------------------------------------------------------------------
    # 3. MÉTODO ESTÁTICO (Usa '@staticmethod': no necesita ni 'self' ni 'cls')
    # --------------------------------------------------------------------------
    @staticmethod
    def es_dia_laboral(numero_dia_semana: int) -> bool:
        """
        Función utilitaria: Recibe un día (1=Lunes ... 7=Domingo) y dice si es laboral.
        Está dentro de Empleado porque tiene relación lógica, pero no requiere datos del empleado.
        """
        return 1 <= numero_dia_semana <= 5


# --- Probando los 3 tipos de métodos ---

# A) Creamos un empleado de la forma normal (__init__) y otro usando el @classmethod:
emp1 = Empleado("Valeria Soto", "Desarrolladora", 2000.0)
emp2 = Empleado.desde_texto_guiones("Marcos Díaz - Analista Datos - 1800")

print("--- Estado inicial ---")
print(emp1.obtener_ficha())
print(emp2.obtener_ficha())

# B) Usamos un @classmethod para cambiar el aumento de toda la empresa al 10% (1.10):
Empleado.fijar_nuevo_aumento_anual(1.10)

# C) Aplicamos el método de instancia a ambos empleados:
emp1.aplicar_aumento_individual()
emp2.aplicar_aumento_individual()

print("\n--- Después de aplicar aumento del 10% ---")
print(emp1.obtener_ficha())
print(emp2.obtener_ficha())

# D) Llamamos al @staticmethod directamente desde la clase:
print(f"\n¿El sábado (día 6) es laboral?: {Empleado.es_dia_laboral(6)}")
```

### 4.2 Encapsulamiento y Propiedades Inteligentes (`@property` y `@setter`)

En muchos lenguajes antiguos se programaban métodos llamados `obtener_precio()` (*getter*) y `cambiar_precio(nuevo)` (*setter*). En Python tenemos algo mucho más limpio y elegante: el decorador **`@property`**.

#### Niveles de privacidad de atributos en Python:
1. **Público (`self.nombre`)**: Se puede leer y modificar libremente desde fuera.
2. **Protegido (`self._precio`)**: Un guion bajo al inicio indica por convención: *"Este atributo es de uso interno de la clase y sus subclases; no lo modifiques directamente desde fuera"*.
3. **Privado (`self.__clave_secreta`)**: Doble guion bajo al inicio activa el *Name Mangling* (Python le cambia el nombre internamente a `_NombreClase__clave_secreta` para evitar que se sobreescriba por accidente).

#### ¿Cómo funcionan `@property` y `@nombre.setter`?
* **`@property` (Getter)**: Convierte un método en un **atributo de lectura**. Puedes calcular un valor al vuelo o devolver un atributo protegido escribiendo simplemente `objeto.precio` (¡sin paréntesis!).
* **`@nombre_propiedad.setter` (Setter)**: Se ejecuta automáticamente cuando alguien escribe `objeto.precio = 500`. Te permite **validar el dato** antes de guardarlo e impedir valores inválidos (como precios negativos).

```python
# Ejemplo 4.2: Uso de @property y @setter para validación automática y atributos calculados

class TermostatoInteligente:
    def __init__(self, ubicacion: str, temp_celsius: float = 22.0):
        self.ubicacion = ubicacion
        # Usamos el setter desde el inicio para que también valide el valor inicial:
        self.temperatura = temp_celsius

    # 1. GETTER (@property): Permite leer 'termostato.temperatura' como si fuera un atributo normal
    @property
    def temperatura(self) -> float:
        return self._temperatura_interna

    # 2. SETTER (@temperatura.setter): Se activa cuando hacemos 'termostato.temperatura = nuevo_valor'
    @temperatura.setter
    def temperatura(self, nuevo_valor: float):
        if not (10.0 <= nuevo_valor <= 35.0):
            raise ValueError(f"Temperatura peligrosa ({nuevo_valor}°C). Debe estar entre 10°C y 35°C.")
        self._temperatura_interna = float(nuevo_valor)

    # 3. PROPIEDAD SOLO DE LECTURA (Calculada automáticamente al vuelo)
    @property
    def temperatura_fahrenheit(self) -> float:
        """Convierte automáticamente los grados Celsius a Fahrenheit sin guardar otro atributo."""
        return round((self._temperatura_interna * 9 / 5) + 32, 1)


# --- Prueba de @property y validación automática ---
sala = TermostatoInteligente("Sala Principal", 24.0)
print(f"Ubicación: {sala.ubicacion}")
print(f"Temperatura actual: {sala.temperatura}°C (Equivale a {sala.temperatura_fahrenheit}°F)")

# Modificamos como si fuera una variable normal (internamente se ejecuta el @temperatura.setter):
sala.temperatura = 19.5
print(f"Nueva temperatura : {sala.temperatura}°C (Equivale a {sala.temperatura_fahrenheit}°F)")

# Intentemos asignar un valor inválido (ej. 50°C) para ver cómo el setter protege al objeto:
try:
    sala.temperatura = 50.0
except ValueError as error:
    print(f"\n[PROTECCIÓN ACTIVADA] No se permitió el cambio: {error}")
```

### 4.3 Métodos Especiales o "Mágicos" (*Dunder Methods*: `__metodo__`)

¿Alguna vez te preguntaste por qué puedes hacer `"hola" + " mundo"`, `len([1, 2, 3])` o `print(mi_objeto)`?
Porque detrás de cada operador de Python (`+`, `-`, `==`, `<`, `len()`, `[]`, `in`, `print()`) hay un **Método Especial con doble guion bajo** (*Dunder = Double UNDERscore*).

Cuando programas estos métodos dentro de tu clase, le enseñas a Python cómo deben comportarse tus objetos con el lenguaje nativo:

| Método Especial | ¿Cuándo lo llama Python automáticamente? | ¿Para qué sirve programarlo? |
| :--- | :--- | :--- |
| `__init__(self, ...)` | Al crear el objeto: `obj = MiClase()` | Inicializar los atributos del objeto. |
| `__str__(self)` | Al usar `print(obj)` o `str(obj)` | Mostrar un texto **amigable para el usuario final**. |
| `__repr__(self)` | Al inspeccionar en consola o usar `repr(obj)` | Mostrar una representación **técnica exacta para programadores**. |
| `__len__(self)` | Al llamar a `len(obj)` | Devolver el tamaño o cantidad de elementos de nuestro objeto. |
| `__eq__(self, otro)` | Al comparar con `obj1 == obj2` | Definir cuándo dos objetos se consideran iguales por su contenido. |
| `__lt__(self, otro)` | Al comparar con `obj1 < obj2` o usar `sorted()` | Permitir ordenar objetos de menor a mayor automáticamente. |
| `__add__(self, otro)` | Al sumar con `obj1 + obj2` | Definir qué significa "sumar" dos objetos de nuestra clase. |
| `__getitem__(self, i)` | Al usar corchetes `obj[indice]` o iterar en un `for` | Hacer que nuestro objeto funcione como una lista o diccionario. |
| `__contains__(self, x)`| Al preguntar `elemento in obj` | Personalizar la búsqueda con el operador `in`. |
| `__call__(self, ...)` | Al llamar al objeto como función: `obj()` | Permitir que una instancia se ejecute igual que una función. |

```python
# Ejemplo 4.3: Programando Métodos Mágicos (Dunder Methods) en una clase CarritoDeCompras

class CarritoDeCompras:
    def __init__(self, cliente: str):
        self.cliente = cliente
        self.items = []  # Lista de tuplas: (nombre_producto, precio)

    def agregar(self, producto: str, precio: float):
        self.items.append((producto, precio))

    @property
    def total(self) -> float:
        return round(sum(precio for _, precio in self.items), 2)

    # 1. __str__: Para que print(carrito) muestre un resumen legible
    def __str__(self) -> str:
        return f"Carrito de {self.cliente} ({len(self)} productos) -> Total: ${self.total:.2f}"

    # 2. __repr__: Para depuración técnica
    def __repr__(self) -> str:
        return f"CarritoDeCompras(cliente='{self.cliente}', items={self.items})"

    # 3. __len__: Permite usar len(carrito)
    def __len__(self) -> int:
        return len(self.items)

    # 4. __getitem__: Permite acceder por índice carrito[0] y recorrer el carrito con un ciclo 'for'
    def __getitem__(self, indice: int):
        return self.items[indice]

    # 5. __contains__: Permite usar ' "Laptop" in carrito '
    def __contains__(self, nombre_buscado: str) -> bool:
        return any(prod.lower() == nombre_buscado.lower() for prod, _ in self.items)

    # 6. __add__: Permite unir dos carritos con el operador '+' (carrito1 + carrito2)
    def __add__(self, otro_carrito):
        if not isinstance(otro_carrito, CarritoDeCompras):
            raise TypeError("Solo puedes sumar dos objetos CarritoDeCompras.")
        nuevo = CarritoDeCompras(f"{self.cliente} & {otro_carrito.cliente}")
        nuevo.items = self.items + otro_carrito.items
        return nuevo

    # 7. __eq__: Permite comparar si dos carritos valen lo mismo (carrito1 == carrito2)
    def __eq__(self, otro_carrito) -> bool:
        if not isinstance(otro_carrito, CarritoDeCompras):
            return False
        return self.total == otro_carrito.total


# --- Demostración del poder de los Métodos Dunder ---
c1 = CarritoDeCompras("Sofía")
c1.agregar("Teclado", 50.0)
c1.agregar("Mouse", 30.0)

c2 = CarritoDeCompras("Mateo")
c2.agregar("Audífonos", 80.0)

print("1. Uso de print() -> llama a __str__      :", c1)
print("2. Uso de len()   -> llama a __len__      :", len(c1), "productos")
print("3. Indexación [0] -> llama a __getitem__  :", c1[0])
print("4. Operador 'in'  -> llama a __contains__ :", "¿Está 'Mouse' en c1?", "mouse" in c1)
print("5. Igualdad '=='  -> llama a __eq__       :", "¿c1 y c2 tienen el mismo total?", c1 == c2)

# ¡Sumamos ambos carritos usando simplemente el signo '+'!
carrito_familiar = c1 + c2
print("6. Suma con '+'   -> llama a __add__      :", carrito_familiar)
```

## 5. Herencia, Polimorfismo, Composición y Clases Abstractas

Cuando construyes aplicaciones reales con múltiples clases, estas se relacionan entre sí mediante cuatro conceptos clave:

### 5.1 Herencia y `super()` (Relación *"Es un..."*)
* Permite crear una **Clase Hija (Subclase)** a partir de una **Clase Padre (Superclase)** para heredar todos sus atributos y métodos sin repetir código.
* La función **`super()`** nos permite llamar a los métodos de la clase padre (por ejemplo, `super().__init__(...)` para que el padre inicialice los atributos comunes antes de agregar los atributos propios del hijo).
* **Sobreescritura de métodos (*Method Overriding*)**: La clase hija puede redefinir un método del padre con el mismo nombre para adaptar su comportamiento.

### 5.2 Polimorfismo
* Significa *"muchas formas"*. Si varias clases distintas (`PagoTarjeta`, `PagoPayPal`, `PagoTransferencia`) tienen un método con el mismo nombre (`procesar_pago()`), nuestro programa puede recorrer una lista de pagos y llamar a `pago.procesar_pago()` sin importar de qué tipo específico sea cada uno.

### 5.3 Clases Base Abstractas (`ABC` y `@abstractmethod`)
* ¿Qué pasa si quieres obligar a que **cualquier clase hija** programe obligatoriamente ciertos métodos?
* Usamos el módulo estándar `abc` (*Abstract Base Classes*) y el decorador **`@abstractmethod`**. Si alguien crea una subclase y olvida programar ese método, Python detendrá el programa inmediatamente con un error claro.

```python
# Ejemplo 5.1: Herencia con super(), Polimorfismo y Clases Abstractas (@abstractmethod)

from abc import ABC, abstractmethod

# 1. CLASE PADRE ABSTRACTA (Plantilla obligatoria para cualquier método de notificación)
class Notificador(ABC):
    def __init__(self, destinatario: str):
        self.destinatario = destinatario

    # Método común heredado por todas las clases hijas
    def formatear_encabezado(self) -> str:
        return f"[Para: {self.destinatario}]"

    # MÉTODO ABSTRACTO: Obliga a todas las clases hijas a programar su propio método 'enviar()'
    @abstractmethod
    def enviar(self, mensaje: str) -> str:
        pass


# 2. CLASES HIJAS (Heredan de Notificador y usan super())
class NotificadorEmail(Notificador):
    def __init__(self, destinatario: str, asunto: str):
        super().__init__(destinatario)  # Llama al __init__ del padre
        self.asunto = asunto            # Atributo exclusivo de Email

    # Implementación obligatoria del método abstracto (Sobreescritura / Polimorfismo)
    def enviar(self, mensaje: str) -> str:
        return f"EMAIL {self.formatear_encabezado()} | Asunto: '{self.asunto}' -> Contenido: {mensaje}"


class NotificadorSMS(Notificador):
    def __init__(self, destinatario: str, codigo_pais: str = "+58"):
        super().__init__(destinatario)
        self.codigo_pais = codigo_pais

    def enviar(self, mensaje: str) -> str:
        mensaje_corto = mensaje[:40]  # Los SMS recortan mensajes largos
        return f"SMS [{self.codigo_pais} {self.destinatario}] -> Texto: {mensaje_corto}"


# --- Demostración de Polimorfismo ---
bandeja_salida = [
    NotificadorEmail("cliente@correo.com", "Factura Disponible"),
    NotificadorSMS("412-5551234"),
]

print("--- Envío polimórfico (distintas clases responden al mismo método .enviar()) ---")
for canal in bandeja_salida:
    print(" ", canal.enviar("Tu pedido #9042 ha sido enviado con éxito."))
```

### 5.4 Herencia vs. Composición: ¿Cuándo usar cada una?
Un consejo de oro en ingeniería de software es: ***"Prefiere la composición sobre la herencia cuando no exista una relación jerárquica estricta"***.

* **Usa Herencia (*"A es un tipo de B"*)**: Un `Perro` **es un** `Animal`; un `NotificadorEmail` **es un** `Notificador`.
* **Usa Composición (*"A tiene un B adentro"*)**: Un `Computador` **no es un** `Procesador`, sino que **tiene un** objeto `Procesador` y **tiene una** `MemoriaRAM` guardados en sus atributos. La composición es mucho más flexible y fácil de modificar.

```python
# Ejemplo 5.2: Composición en acción ("Tiene un...")

class Motor:
    def __init__(self, caballos_fuerza: int, tipo: str = "Eléctrico"):
        self.caballos_fuerza = caballos_fuerza
        self.tipo = tipo
        self.encendido = False

    def arrancar(self) -> str:
        self.encendido = True
        return f"Motor {self.tipo} de {self.caballos_fuerza} HP encendido silenciosamente."


class Vehiculo:
    def __init__(self, marca: str, modelo: str, motor: Motor):
        self.marca = marca
        self.modelo = modelo
        # COMPOSICIÓN: El atributo 'self.motor' guarda un objeto completo de la clase Motor
        self.motor = motor

    def iniciar_marcha(self):
        estado_motor = self.motor.arrancar()  # Delegamos la acción al método del objeto Motor
        print(f"[{self.marca} {self.modelo}] -> {estado_motor}")


mi_motor = Motor(caballos_fuerza=280, tipo="Híbrido")
mi_auto = Vehiculo("Toyota", "Corolla Cross", motor=mi_motor)
mi_auto.iniciar_marcha()
```

## 6. Clases Modernas con `@dataclass` y Consejos Profesionales de Desarrollo

### 6.1 El Decorador `@dataclass`: Clases Limpias en Segundos
Muchas veces creamos clases cuyo objetivo principal es guardar datos estructurados (como un `Producto`, una `Coordenada` o un `RegistroVenta`) junto con algunos métodos.
Escribir manualmente `self.a = a`, `self.b = b`, `__repr__` y `__eq__` una y otra vez es repetitivo.

Desde Python 3.7 existe el decorador **`@dataclass`** (del módulo estándar `dataclasses`):
* Solo declaras los atributos con sus tipos (`nombre: str`).
* **`@dataclass` escribe automáticamente por ti** los métodos `__init__`, `__repr__` y `__eq__`.
* Y por supuesto, **puedes seguir programando todos los métodos de instancia, `@property`, `@classmethod` y `@staticmethod` que quieras** dentro de ella.

```python
# Ejemplo 6: Uso de @dataclass combinada con métodos personalizados y @property

from dataclasses import dataclass, field
from typing import List

@dataclass
class CursoOnline:
    titulo: str
    instructor: str
    precio_usd: float
    calificaciones: List[int] = field(default_factory=list)  # Forma segura de crear listas vacías por defecto

    # Podemos programar cualquier método o @property igual que en una clase tradicional:
    def agregar_calificacion(self, estrellas: int):
        if not (1 <= estrellas <= 5):
            raise ValueError("La calificación debe estar entre 1 y 5 estrellas.")
        self.calificaciones.append(estrellas)

    @property
    def promedio_estrellas(self) -> float:
        if not self.calificaciones:
            return 0.0
        return round(sum(self.calificaciones) / len(self.calificaciones), 2)


# ¡No tuvimos que escribir __init__ ni __repr__!
curso = CursoOnline(titulo="Python Profesional", instructor="Elena Gómez", precio_usd=49.99)
curso.agregar_calificacion(5)
curso.agregar_calificacion(4)
curso.agregar_calificacion(5)

print("Representación automática (__repr__) de @dataclass:")
print(" ", curso)
print(f"Promedio calculado con nuestro método @property: {curso.promedio_estrellas} ⭐")
```

### 6.2 Consejos Profesionales para Programar Clases y Métodos en el Mundo Real

1. **¿Cuándo crear una Clase y cuándo usar solo Funciones?**
   * **Usa Funciones** cuando solo necesitas recibir datos, transformarlos y devolver un resultado sin guardar memoria entre llamadas (ej. `convertir_divisa(monto, tasa)`).
   * **Usa una Clase** cuando tienes un **estado (datos)** que debe mantenerse unido y evolucionar a través de varias operaciones (ej. una `SesionUsuario`, un `CarritoDeCompras`, un `PersonajeJuego` o una `ConexionBaseDeDatos`).
2. **El truco para detectar que una clase es innecesaria**:
   * Si programas una clase que solo tiene `__init__` y **un único método más**, casi siempre es más simple y legible reemplazarla por una función normal.
3. **Mantén el método `__init__` ligero y rápido**:
   * `__init__` solo debe asignar y validar atributos iniciales. Evita hacer descargas de internet, consultas pesadas a bases de datos o abrir archivos gigantes automáticamente dentro de `__init__`; para eso es mejor crear un método explícito como `objeto.conectar()` o un `@classmethod` como `Clase.cargar_desde_archivo()`.
4. **Usa `@property` para mantener tu código preparado para el futuro**:
   * En Python puedes empezar con un atributo público simple (`self.precio = 100`). Si meses después el negocio exige validar que el precio nunca sea menor a 0, lo conviertes en `@property` + `@precio.setter` sin tener que cambiar ni una sola línea en el resto de tu programa donde se usaba `producto.precio`.
5. **Siempre programa `__repr__` (o usa `@dataclass`)**:
   * Cuando estés buscando un error a las 11 de la noche, ver `<__main__.Cliente object at 0x7f8b...>` en la consola no te ayudará en nada. En cambio, si programas `__repr__` para que muestre `Cliente(id=42, nombre='Ana', activo=True)`, encontrarás los errores en segundos.
6. **Métodos cortos y cohesivos**:
   * Cada método dentro de una clase debe trabajar con los atributos de su propia clase. Si notas que un método no usa `self` en ningún momento, conviértelo en `@staticmethod` o sácalo como una función auxiliar independiente.
