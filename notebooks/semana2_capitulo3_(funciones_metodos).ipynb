## 1. Diferencia Conceptual: ¿Qué es una Función y qué es un Método?

Aunque ambos son bloques de código reutilizables que realizan una tarea específica, existe una diferencia estructural clave en Python:

* **Función (`Function`)**: Es un bloque de código **independiente** definido fuera de una clase (a nivel de módulo o dentro de otra función). Se llama directamente por su nombre pasándole los datos con los que debe trabajar: `nombre_funcion(datos)`.
* **Método (`Method`)**: Es una función que **pertenece a un objeto o a una clase**. Está vinculada a los datos de ese objeto y se invoca utilizando la **notación de punto**: `objeto.nombre_metodo(argumentos)`. Implícitamente, el método conoce al objeto que lo llamó.

| Característica | Función | Método |
| :--- | :--- | :--- |
| **Dependencia** | Independiente (existe por sí sola en el módulo) | Asociada a una clase o instancia de un objeto |
| **Forma de llamarla** | `funcion(arg1, arg2)` | `objeto.metodo(arg1, arg2)` |
| **Acceso al estado** | Solo accede a lo que recibe por parámetro (o alcance global) | Accede y puede modificar el estado interno del objeto (`self`) |
| **Ejemplo nativo en Python** | `len(lista)`, `sum(numeros)`, `sorted(datos)` | `texto.upper()`, `lista.append(5)`, `diccionario.get("id")` |

````python
# Ejemplo 1: Comparación directa entre Función y Método

mensaje = "  hola mundo pythonico  "

# 1. Uso de una FUNCIÓN integrada (independiente del tipo, recibe el objeto como argumento)
longitud_inicial = len(mensaje)
print(f"Longitud calculada con la función len(): {longitud_inicial} caracteres")

# 2. Uso de MÉTODOS del objeto 'str' (se invocan con un punto sobre el objeto)
mensaje_limpio = mensaje.strip()          # Método que elimina espacios en los extremos
mensaje_titulo = mensaje_limpio.title()   # Método que capitaliza cada palabra

print(f"Resultado usando métodos (.strip y .title): '{mensaje_titulo}'")
print(f"Nueva longitud con función len(): {len(mensaje_titulo)} caracteres")
````

## 2. Anatomía y Fundamentos de las Funciones

Para crear una función propia en Python utilizamos la palabra reservada `def`. Su estructura fundamental se compone de:

1. **Encabezado (`Header`)**: `def nombre_funcion(parametro1, parametro2):`
   * El nombre debe escribirse en formato `snake_case` (minúsculas separadas por guiones bajos) según la guía de estilo oficial **PEP 8**.
2. **Docstring (Cadena de documentación)**: Texto entre triples comillas `"""..."""` justo en la primera línea dentro de la función. Explica qué hace la función, qué parámetros recibe y qué devuelve. Es consultable mediante `help(nombre_funcion)` o `nombre_funcion.__doc__`.
3. **Cuerpo (`Body`)**: Bloque de código indentado (4 espacios) que contiene la lógica.
4. **Sentencia de retorno (`return`)**: Devuelve el resultado al lugar donde se llamó la función y **finaliza inmediatamente** su ejecución. Si una función no tiene `return` (o tiene un `return` vacío), Python devuelve automáticamente `None`.

### Distinción clave: Parámetros vs. Argumentos
* **Parámetros**: Son los nombres de las variables declaradas en la **definición** de la función (`def sumar(a, b):` -> `a` y `b` son parámetros).
* **Argumentos**: Son los valores reales que enviamos al **invocar** la función (`sumar(10, 25)` -> `10` y `25` son argumentos).

### Distinción clave: `print()` vs. `return`
Un error común al aprender Python es confundir mostrar un dato en pantalla (`print`) con devolver un dato (`return`):
* `print()` solo muestra texto en la consola para el humano, pero **no entrega ningún valor** al programa (devuelve `None`).
* `return` entrega el dato calculado para que pueda guardarse en una variable, usarse en una operación matemática o pasarse a otra función.

````python
# Ejemplo 2.1: Anatomía completa de una función y diferencia entre print() y return

def calcular_precio_final(precio_base, porcentaje_impuesto):
    """
    Calcula el precio final de un producto aplicando el porcentaje de impuesto.

    Parámetros:
        precio_base (float): Costo original del producto sin impuestos.
        porcentaje_impuesto (float): Impuesto expresado en porcentaje (ej. 16 para 16%).

    Retorna:
        float: Precio final redondeado a 2 decimales.
    """
    impuesto = precio_base * (porcentaje_impuesto / 100)
    total = round(precio_base + impuesto, 2)
    return total


# Llamamos a la función pasando ARGUMENTOS (150.0 y 16)
total_laptop = calcular_precio_final(150.0, 16)
print(f"Precio final devuelto con return: ${total_laptop}")

# Gracias a 'return', podemos reutilizar el resultado en otros cálculos:
total_dos_laptops = total_laptop * 2
print(f"Precio por 2 unidades: ${total_dos_laptops}")

# Consultamos el docstring de nuestra función:
print("\n--- Documentación interna (__doc__) ---")
print(calcular_precio_final.__doc__.strip())
````

### Retorno de Múltiples Valores (Empaquetado y Desempaquetado)
En Python, una función puede devolver varios resultados separados por comas. Internamente, Python los empaqueta en una **tupla** (`tuple`), la cual podemos desempaquetar directamente en múltiples variables al llamar a la función.

````python
# Ejemplo 2.2: Retorno de múltiples valores y desempaquetado (unpacking)

def analizar_calificaciones(notas):
    """Calcula el promedio, la nota mínima, la nota máxima y la cantidad de aprobados (>= 60)."""
    if not notas:
        return 0, 0, 0, 0

    promedio = round(sum(notas) / len(notas), 2)
    minima = min(notas)
    maxima = max(notas)
    aprobados = sum(1 for n in notas if n >= 60)

    # Retornamos 4 valores separados por comas (se empaquetan como tupla)
    return promedio, minima, maxima, aprobados


calificaciones_curso = [85, 92, 58, 74, 100, 45, 88]

# Desempaquetado directo en 4 variables independientes:
prom, nota_min, nota_max, total_aprobados = analizar_calificaciones(calificaciones_curso)

print(f"Promedio del grupo : {prom}")
print(f"Rango de notas     : Mínima = {nota_min} | Máxima = {nota_max}")
print(f"Alumnos aprobados  : {total_aprobados} de {len(calificaciones_curso)}")
````

## 3. Tipos de Argumentos y Firmas Avanzadas

Python posee uno de los sistemas de paso de argumentos más flexibles y expresivos de todos los lenguajes de programación.

### 3.1 Argumentos Posicionales, Keyword (por nombre) y Valores por Defecto
1. **P
osicionales**: Se asignan según el orden estricto en que se pasan.
2. **Por nombre (`Keyword Arguments`)**: Se especifica explícitamente `nombre_parametro=valor`. Permiten cambiar el orden y hacen que el llamado sea mucho más legible.
   * *Regla de sintaxis*: Los argumentos posicionales siempre deben ir **antes** que los argumentos por nombre al llamar a una función.
3. **Valores por defecto (`Default Parameters`)**: Se definen en el encabezado (`def conectar(puerto=5432):`). Si el usuario no envía ese argumento, toma el valor predeterminado.

````python
# Ejemplo 3.1: Argumentos posicionales, por nombre y valores por defecto

def crear_perfil_usuario(nombre, rol="estudiante", activo=True, nivel=1):
    """Genera un diccionario con la configuración de un usuario."""
    return {
        "nombre": nombre,
        "rol": rol,
        "activo": activo,
        "nivel": nivel
    }

# 1. Solo con el argumento obligatorio (usa todos los valores por defecto)
u1 = crear_perfil_usuario("Ana Gómez")

# 2. Combinando posicionales y argumentos por nombre (en cualquier orden para los keyword)
u2 = crear_perfil_usuario("Carlos Ruiz", nivel=5, rol="instructor")

print("Usuario 1 (valores por defecto):", u1)
print("Usuario 2 (argumentos keyword) :", u2)
````

### 3.2 ¡Cuidado! La Trampa de los Argumentos Mutables por Defecto
Este es uno de los errores más famosos en entrevistas técnicas y bugs en producción en Python:

* **El problema**: Los valores por defecto se evalúan **una única vez** cuando Python lee el archivo y crea la función (`def`), **NO** cada vez que se llama a la función.
* Si colocas una lista o diccionario como valor por defecto (`def agregar_item(item, carrito=[])`), ¡todas las llamadas que no pasen `carrito` compartirán exactamente **la misma lista en memoria**!
* **La solución profesional**: Usar `None` como valor por defecto y crear la lista/diccionario nuevo dentro de la función.

````python
# Ejemplo 3.2: La trampa de los mutables por defecto vs. Patrón correcto con None

# INCORRECTO: Usar una lista mutable [] como valor por defecto
def agregar_tarea_mal(tarea, lista_tareas=[]):
    lista_tareas.append(tarea)
    return lista_tareas

print("--- Comportamiento INCORRECTO (comparte la misma lista entre llamadas) ---")
print("Llamada 1:", agregar_tarea_mal("Estudiar Python"))
print("Llamada 2:", agregar_tarea_mal("Hacer ejercicio"))  # ¡Aparece también la tarea de la llamada 1!


# CORRECTO: Usar None como valor centinela e inicializar adentro
def agregar_tarea_bien(tarea, lista_tareas=None):
    if lista_tareas is None:
        lista_tareas = []
    lista_tareas.append(tarea)
    return lista_tareas

print("\n--- Comportamiento CORRECTO (cada llamada tiene su propia lista nueva) ---")
print("Llamada 1:", agregar_tarea_bien("Estudiar Python"))
print("Llamada 2:", agregar_tarea_bien("Hacer ejercicio"))
````

### 3.3 Argumentos de Longitud Variable: `*args` y `**kwargs`
En ocasiones no sabemos de antemano cuántos argumentos necesitará pasar el usuario:

* **`*args` (Arguments posicionales variables)**: El asterisco simple `*` agrupa cualquier cantidad de argumentos posicionales adicionales dentro de una **tupla** (`tuple`).
* **`**kwargs` (Keyword Arguments variables)**: El doble asterisco `**` agrupa cualquier cantidad de argumentos pasados por nombre (`clave=valor`) dentro de un **diccionario** (`dict`).

*(Nota: Los nombres `args` y `kwargs` son una convención universal entre programadores de Python; lo que tiene efecto sintáctico son los operadores `*` y `**`).*

````python
# Ejemplo 3.3: Uso combinado de parámetros fijos, *args y **kwargs

def registrar_pedido(cliente, *platillos, descuento=0.0, **detalles_entrega):
    """
    Registra un pedido de restaurante:
    - cliente: parámetro obligatorio
    - *platillos: tupla con todos los platillos pedidos
    - descuento: parámetro con valor por defecto
    - **detalles_entrega: diccionario con metadatos adicionales (dirección, teléfono, notas)
    """
    print(f"=== Pedido de: {cliente} ===")
    print(f"Platillos ({len(platillos)}): {', '.join(platillos)}")
    print(f"Descuento aplicado: {descuento * 100:.0f}%")
    if detalles_entrega:
        print("Detalles de entrega:")
        for clave, valor in detalles_entrega.items():
            print(f"  - {clave}: {valor}")
    print()


registrar_pedido(
    "Elena Torres",
    "Pizza Margarita", "Ensalada César", "Limonada",  # Capturados en *platillos (tupla)
    descuento=0.15,                                   # Parámetro keyword explícito
    direccion="Av. Libertador 402",                   # Capturados en **detalles_entrega (dict)
    metodo_pago="Tarjeta",
    propina_incluida=True
)
````

### 3.4 Control Estricto de Firma: `/` (Solo Posicional) y `*` (Solo por Nombre)
A partir de Python 3.8, puedes diseñar APIs más seguras obligando a que ciertos parámetros se pasen exclusivamente por posición o exclusivamente por nombre:

```python
def funcion(pos1, pos2, /, estandar1, estandar2, *, kw1, kw2):
```
* Todo lo que esté **a la izquierda de `/`** debe pasarse **solo por posición** (no se permite usar su nombre al llamar).
* Todo lo que esté **entre `/` y `*`** puede pasarse por posición o por nombre.
* Todo lo que esté **a la derecha de `*`** debe pasarse **obligatoriamente por nombre (`keyword-only`)**. Esto es excelente para banderas booleanas o configuraciones críticas.

````python
# Ejemplo 3.4: Parámetros solo-posicionales (/) y solo-por-nombre (*)

def transferir_fondos(cuenta_origen, cuenta_destino, /, monto, *, comision=0.0, confirmar_envio=False):
    """
    - cuenta_origen y cuenta_destino: SOLO posicionales (antes de /)
    - monto: posicional o por nombre
    - comision y confirmar_envio: SOLO por nombre (después de *), obligando a que el código sea explícito
    """
    if not confirmar_envio:
        return "Transferencia cancelada: se requiere confirmar_envio=True"
    total_debitado = monto + comision
    return f"Transferencia exitosa de {cuenta_origen} a {cuenta_destino} por ${monto:.2f} (Total debitado: ${total_debitado:.2f})"


# Llamada válida y sumamente clara:
resultado = transferir_fondos("CTA-001", "CTA-999", 500.0, comision=2.50, confirmar_envio=True)
print(resultado)

# Si intentáramos llamar transferir_fondos("CTA-001", "CTA-999", 500.0, 2.50, True)
# Python lanzaría un TypeError porque 'comision' y 'confirmar_envio' exigen pasarse por nombre.
````

## 4. Alcance de Variables (Regla LEGB), Ciudadanos de Primera Clase y Clausuras

### 4.1 La Regla LEGB de Resolución de Nombres
Cuando usas una variable dentro de una función, Python la busca siguiendo estrictamente el orden **LEGB**:

1. **L — Local**: Variables definidas dentro de la función actual (incluyendo sus parámetros). Nacen cuando la función se llama y se destruyen cuando termina.
2. **E — Enclosing (Contenedor)**: Si hay una función dentro de otra función (funciones anidadas), busca en el ámbito local de la función exterior.
3. **G — Global**: Variables definidas en el nivel principal del archivo/módulo (fuera de toda función).
4. **B — Built-in (Integrado)**: Nombres predefinidos del lenguaje Python (`len`, `range`, `Exception`, `True`, etc.).

> **Regla de Oro sobre `global` y `nonlocal`**:
> * Dentro de una función puedes **leer** una variable global sin problema. Pero si intentas **reasignarla** (`contador = contador + 1`), Python asume que estás creando una variable *local* nueva.
> * Aunque existen las palabras clave `global` (para modificar variables globales) y `nonlocal` (para modificar variables de una función contenedora), **modificar variables globales desde funciones es una mala práctica** porque hace el código impredecible y difícil de probar. En cambio, `nonlocal` sí es útil dentro de **Clausuras (Closures)**.

````python
# Ejemplo 4.1: Demostración práctica de la regla LEGB y funciones como objetos (Closures)

impuesto_global = 0.16  # Ámbito GLOBAL (G)

def crear_contador_ventas(tienda):
    """
    Fábrica de funciones (Closure):
    Recuerda el nombre de la 'tienda' y mantiene su propio estado 'total_acumulado'
    en el ámbito ENCLOSING (E) sin usar variables globales.
    """
    total_acumulado = 0.0  # Ámbito ENCLOSING (E)

    def registrar_venta(monto_sin_iva):
        nonlocal total_acumulado  # Permite modificar la variable del ámbito Enclosing

        # 'monto_con_iva' está en el ámbito LOCAL (L)
        # 'impuesto_global' se lee del ámbito GLOBAL (G)
        # 'round' proviene del ámbito BUILT-IN (B)
        monto_con_iva = round(monto_sin_iva * (1 + impuesto_global), 2)
        total_acumulado = round(total_acumulado + monto_con_iva, 2)
        return f"[{tienda}] Venta: ${monto_con_iva} | Acumulado en caja: ${total_acumulado}"

    return registrar_venta  # Devolvemos la función interna (sin paréntesis)


# Creamos dos cajas registradoras independientes:
caja_norte = crear_contador_ventas("Sucursal Norte")
caja_sur = crear_contador_ventas("Sucursal Sur")

print(caja_norte(100.0))
print(caja_norte(50.0))
print(caja_sur(200.0))  # Sucursal Sur tiene su propio estado independiente
````

## 5. Funciones Especiales y Herramientas Funcionales

En Python las funciones son **ciudadanos de primera clase**: pueden asignarse a variables, guardarse en listas/diccionarios, pasarse como argumentos a otras funciones y devolverse como resultado. Esto habilita herramientas muy potentes:

### 5.1 Funciones Anónimas (`lambda`)
Una función `lambda` es una función pequeña, de **una sola expresión**, que no necesita nombre (`def`).
* **Sintaxis**: `lambda parametros: expresion_de_retorno`
* **¿Cuándo usarla?**: Cuando una función como `sorted()`, `min()`, `max()` o `map()` requiere una función pequeña como argumento `key` que solo usarás en ese instante.
* **¿Cuándo NO usarla?**: Si la lógica es compleja o si vas a asignarla a una variable (`f = lambda x: ...`). La guía PEP 8 indica que si necesitas darle nombre, siempre debes usar `def`.

````python
# Ejemplo 5.1: Uso práctico de lambda para ordenar estructuras de datos complejas

productos = [
    {"nombre": "Teclado Mecánico", "precio": 85.0, "stock": 12},
    {"nombre": "Monitor 27 pulgadas", "precio": 290.0, "stock": 4},
    {"nombre": "Mouse Inalámbrico", "precio": 35.0, "stock": 30},
    {"nombre": "Audífonos Estudio", "precio": 120.0, "stock": 8},
]

# Ordenamos la lista de diccionarios de menor a mayor precio usando una función lambda en 'key':
ordenados_por_precio = sorted(productos, key=lambda prod: prod["precio"])

print("Productos ordenados por precio (menor a mayor):")
for p in ordenados_por_precio:
    print(f"  - {p['nombre']:<22} | Precio: ${p['precio']:>6.2f} | Stock: {p['stock']}")

# Obtenemos el producto con mayor stock usando max() + lambda:
mayor_stock = max(productos, key=lambda prod: prod["stock"])
print(f"\nProducto con más unidades en inventario: {mayor_stock['nombre']} ({mayor_stock['stock']} uds.)")
````

### 5.2 `map()`, `filter()` vs. Comprensión de Listas (List Comprehensions)
* `map(funcion, iterable)`: Aplica `funcion` a cada elemento del iterable.
* `filter(funcion_booleana, iterable)`: Filtra los elementos donde la función devuelve `True`.
* **Consejo Pythónico**: Aunque `map` y `filter` son útiles (especialmente cuando ya tienes una función definida), en Python moderno las **List Comprehensions** (`[x for x in lista if condicion]`) suelen preferirse porque son más directas y fáciles de leer.

````python
# Ejemplo 5.2: Comparación entre map/filter y List Comprehensions

precios_usd = [10, 25, 60, 15, 80, 120]

# Objetivo: Filtrar precios >= 50 y aplicarles un 10% de descuento.

# Enfoque 1: Con map() y filter()
precios_desc_funcional = list(map(lambda p: round(p * 0.9, 2), filter(lambda p: p >= 50, precios_usd)))

# Enfoque 2: Con List Comprehension (Más legible y pythónico)
precios_desc_comprehension = [round(p * 0.9, 2) for p in precios_usd if p >= 50]

print("Resultado con map/filter        :", precios_desc_funcional)
print("Resultado con List Comprehension:", precios_desc_comprehension)
````

### 5.3 Generadores (`yield`) vs. Funciones Tradicionales (`return`)
Cuando trabajas con millones de registros o archivos gigantes, devolver una lista completa con `return` puede agotar la memoria RAM de tu computadora.

* Una función que utiliza **`yield`** en lugar de `return` se convierte en una **Función Generadora**.
* En vez de calcular y devolver todos los elementos de golpe, **produce un valor a la vez** bajo demanda (*evaluación perezosa / lazy evaluation*), "pausando" su estado interno hasta que se le solicita el siguiente valor en un bucle `for` o con `next()`.

````python
# Ejemplo 5.3: Función Generadora con 'yield' para procesamiento eficiente en memoria

import sys

def cuadrados_con_lista(n):
    """Construye y guarda TODOS los elementos en memoria antes de retornar."""
    resultado = []
    for i in range(n):
        resultado.append(i ** 2)
    return resultado

def cuadrados_con_generador(n):
    """Produce cada cuadrado uno por uno al vuelo sin guardar toda la secuencia en RAM."""
    for i in range(n):
        yield i ** 2


lista_100k = cuadrados_con_lista(100_000)
gen_100k = cuadrados_con_generador(100_000)

print(f"Memoria ocupada por la LISTA de 100,000 elementos    : {sys.getsizeof(lista_100k):,} bytes")
print(f"Memoria ocupada por el GENERADOR de 100,000 elementos: {sys.getsizeof(gen_100k):,} bytes")

# Consumimos los primeros 5 valores del generador usando next():
primeros_cinco = [next(gen_100k) for _ in range(5)]
print("Primeros 5 valores producidos por el generador:", primeros_cinco)
````

### 5.4 Decoradores (`@decorator`)
Un **decorador** es una función que recibe otra función como entrada, le añade alguna funcionalidad extra (antes o después de ejecutarla) y devuelve la función mejorada, **sin modificar el código original de la función decorada**.

Se utilizan muchísimo en el mundo real para:
* Medir el tiempo de ejecución de funciones.
* Registrar auditorías/logs de llamadas.
* Verificar permisos o autenticación en frameworks web (`@login_required`, `@app.get("/usuarios")`).
* Guardar en caché resultados costosos (`@functools.lru_cache`).

````python
# Ejemplo 5.4: Creación y uso de un Decorador profesional con @functools.wraps

import time
from functools import wraps

def medir_tiempo(funcion_original):
    """Decorador que mide e imprime cuánto tarda en ejecutarse cualquier función."""
    @wraps(funcion_original)  # Preserva el nombre y docstring de la función original
    def envoltura(*args, **kwargs):
        inicio = time.perf_counter()
        resultado = funcion_original(*args, **kwargs)
        duracion_ms = (time.perf_counter() - inicio) * 1000
        print(f"[LOG TIEMPO] La función '{funcion_original.__name__}' tardó {duracion_ms:.3f} ms")
        return resultado
    return envoltura


@medir_tiempo
def procesar_lote_datos(cantidad):
    """Simula el procesamiento matemático de un lote de números."""
    return sum(i * i for i in range(cantidad))


total = procesar_lote_datos(200_000)
print(f"Resultado del cálculo: {total:,}")
print(f"Nombre preservado gracias a @wraps: {procesar_lote_datos.__name__}")
````

## 6. Métodos en Python a Profundidad (Tipos Nativos y POO)

Como vimos al inicio, un **método** es una función que pertenece a un objeto o clase. En Python todo es un objeto (cadenas, listas, diccionarios, números y tus propias clases), por lo que trabajar con métodos es parte del día a día.

### 6.1 Métodos Esenciales de los Tipos Integrados (`str`, `list`, `dict`)
Un aspecto crucial al usar métodos de objetos nativos es saber **si el método modifica el objeto original (*in-place*) o si devuelve un objeto nuevo**:

* **Cadenas (`str`)**: Son **inmutables**. Ningún método de `str` modifica la cadena original; **siempre devuelven una cadena nueva**.
* **Listas (`list`)**: Son **mutables**. Métodos como `.append()`, `.extend()`, `.sort()` o `.reverse()` modifican la lista *in-place* y **devuelven `None`** (¡no hagas `mi_lista = mi_lista.append(x)`!).
* **Diccionarios (`dict`)**: Son **mutables**. Métodos como `.get(clave, valor_por_defecto)` evitan errores `KeyError` cuando una clave no existe.

````python
# Ejemplo 6.1: Métodos esenciales de str, list y dict (Inmutabilidad vs. Mutación in-place)

# 1. Métodos de cadenas (str) -> Siempre retornan una NUEVA cadena
ruta_archivo = "   reporte_ventas_2026.CSV   "
partes = ruta_archivo.strip().lower().replace(".csv", "").split("_")
print("1. Métodos encadenados en str :", partes)
print("   Unión con .join()          :", "-".join(partes))

# 2. Métodos de listas (list) -> Modifican la lista original in-place
tareas = ["Diseñar base de datos", "Crear API"]
tareas.append("Escribir tests")               # Agrega 1 elemento al final
tareas.extend(["Documentar", "Desplegar"])    # Agrega múltiples elementos de otro iterable
ultima_eliminada = tareas.pop()               # Extrae y retorna el último elemento
print(f"\n2. Lista modificada in-place  : {tareas} (Se extrajo: '{ultima_eliminada}')")

# 3. Métodos seguros de diccionarios (dict)
config = {"host": "localhost", "puerto": 8080}
# .get() devuelve el valor si existe, o el valor por defecto si no existe (sin lanzar KeyError)
timeout = config.get("timeout_segundos", 30)
print(f"\n3. Consulta segura con .get() : host={config.get('host')}, timeout={timeout}s")
````

### 6.2 Los 3 Tipos de Métodos dentro de una Clase (POO)
Cuando diseñas tus propias clases con `class`, puedes definir tres tipos principales de métodos, además de propiedades y métodos mágicos:

1. **Método de Instancia (`def metodo(self, ...):`)**
   * Es el tipo más común. Recibe `self` como primer parámetro, el cual representa al **objeto específico (instancia)** que invocó el método.
   * Puede leer y modificar los atributos de ese objeto (`self.saldo`, `self.nombre`) y llamar a otros métodos del mismo objeto.

2. **Método de Clase (`@classmethod` + `def metodo(cls, ...):`)**
   * Se decora con `@classmethod` y recibe `cls` (la **clase misma**, no una instancia) como primer parámetro.
   * **¿Para qué se usa?**: Su uso estrella en Python es crear **Constructores Alternativos (Factory Methods)**, por ejemplo `Producto.desde_texto("Teclado|85.0")` o `Fecha.hoy()`, además de acceder o modificar atributos compartidos a nivel de clase.

3. **Método Estático (`@staticmethod` + `def metodo(...):`)**
   * Se decora con `@staticmethod` y **no recibe ni `self` ni `cls`**.
   * **¿Para qué se usa?**: Es una función auxiliar normal que vive dentro de la clase por organización lógica (porque tiene relación conceptual con la clase), pero no necesita leer ni modificar ningún dato del objeto ni de la clase.

### 6.3 Métodos Especiales ("Dunder / Magic Methods") y `@property`
* **Métodos Dunder (`__nombre__`)**: Tienen doble guion bajo al inicio y al final (*Double UNDERscore*). Python los invoca automáticamente:
  * `__init__(self, ...)`: Inicializador que configura los atributos al crear el objeto.
  * `__str__(self)`: Representación legible para usuarios finales (cuando haces `print(objeto)`).
  * `__repr__(self)`: Representación técnica inequívoca para desarrolladores y depuración.
* **Propiedades (`@property`)**: Permiten definir un método que se accede como si fuera un atributo normal (`objeto.precio_con_iva` sin paréntesis), ideal para atributos calculados dinámicamente o validaciones (*getters* y *setters* limpios).

````python
# Ejemplo 6.2: Clase completa integrando los 3 tipos de métodos, @property y Dunder Methods

class ProductoInventario:
    # Atributo de clase compartido por todas las instancias
    iva_general = 0.16

    # 1. MÉTODO DUNDER INICIALIZADOR
    def __init__(self, sku: str, nombre: str, precio_neto: float, stock: int = 0):
        if not self.es_sku_valido(sku):
            raise ValueError(f"El SKU '{sku}' no cumple el formato válido (ej. 'PROD-101').")
        self.sku = sku
        self.nombre = nombre
        self.precio_neto = precio_neto
        self.stock = stock

    # 2. MÉTODOS DE INSTANCIA (reciben 'self' y operan sobre un producto específico)
    def reabastecer(self, cantidad: int) -> int:
        """Aumenta el inventario del producto y devuelve el nuevo stock."""
        if cantidad <= 0:
            raise ValueError("La cantidad a reabastecer debe ser positiva.")
        self.stock += cantidad
        return self.stock

    def vender(self, cantidad: int) -> float:
        """Descuenta unidades del stock y retorna el monto total de la venta con IVA."""
        if cantidad > self.stock:
            raise ValueError(f"Stock insuficiente para {self.nombre} (Disponible: {self.stock}).")
        self.stock -= cantidad
        return round(cantidad * self.precio_con_iva, 2)

    # 3. PROPIEDAD CALCULADA (@property: se usa como atributo sin paréntesis)
    @property
    def precio_con_iva(self) -> float:
        """Calcula dinámicamente el precio final incluyendo el IVA de la clase."""
        return round(self.precio_neto * (1 + self.iva_general), 2)

    # 4. MÉTODO DE CLASE (@classmethod: recibe 'cls', ideal como constructor alternativo)
    @classmethod
    def desde_cadena_csv(cls, linea_csv: str):
        """Constructor alternativo: crea un ProductoInventario a partir de un texto 'SKU,Nombre,Precio,Stock'."""
        sku, nombre, precio_str, stock_str = [campo.strip() for campo in linea_csv.split(",")]
        return cls(sku, nombre, float(precio_str), int(stock_str))

    # 5. MÉTODO ESTÁTICO (@staticmethod: no recibe ni 'self' ni 'cls', función utilitaria de validación)
    @staticmethod
    def es_sku_valido(sku: str) -> bool:
        """Valida que el código SKU tenga el formato 'XXXX-NNN'."""
        partes = sku.split("-")
        return len(partes) == 2 and partes[0].isalpha() and partes[1].isdigit()

    # 6. MÉTODOS DUNDER DE REPRESENTACIÓN
    def __str__(self) -> str:
        return f"[{self.sku}] {self.nombre} — Neto: ${self.precio_neto:.2f} (Con IVA: ${self.precio_con_iva:.2f}) | Stock: {self.stock}"

    def __repr__(self) -> str:
        return f"ProductoInventario(sku='{self.sku}', nombre='{self.nombre}', precio_neto={self.precio_neto}, stock={self.stock})"


# --- Demostración práctica ---
# A) Creación mediante el constructor alternativo (@classmethod)
monitor = ProductoInventario.desde_cadena_csv("TECH-205, Monitor UltraWide 34, 450.00, 10")
print("Objeto creado con @classmethod:")
print(" ", monitor)  # Invoca automáticamente __str__

# B) Uso de métodos de instancia (self) y @property
nuevo_stock = monitor.reabastecer(5)
monto_cobrado = monitor.vender(2)
print(f"\nVenta de 2 unidades realizada por: ${monto_cobrado}")
print(f"Stock actualizado tras reabastecer (+5) y vender (-2): {monitor.stock} unidades")
print(f"Acceso a @property precio_con_iva: ${monitor.precio_con_iva}")

# C) Uso directo del @staticmethod sin necesidad de instanciar ningún objeto
print("\nValidación directa con @staticmethod:")
print("  ¿Es válido 'TECH-205'? ->", ProductoInventario.es_sku_valido("TECH-205"))
print("  ¿Es válido '12345'?    ->", ProductoInventario.es_sku_valido("12345"))
````

## 7. ¿Para Qué se Utilizan en la Vida Real y Consejos Profesionales de Desarrollo?

### 7.1 ¿Para qué se utilizan las Funciones y Métodos en proyectos reales?

En el desarrollo profesional de software, análisis de datos e inteligencia artificial, nunca escribimos miles de líneas sueltas en un archivo. Las funciones y métodos son los **bloques de construcción fundamentales** por cinco razones estratégicas:

1. **Principio DRY (*Don't Repeat Yourself* — No te repitas)**:
   * Si necesitas validar un correo electrónico o calcular una métrica financiera en 15 partes del sistema, lo escribes **una sola vez** en una función. Si mañana cambia la regla de negocio, solo modificas esa función y todo el sistema se actualiza sin errores.
2. **Abstracción y Descomposición de Problemas Complejos**:
   * Dividen un problema gigante (ej. *"Procesar pago de carrito de compras"*) en subproblemas pequeños y comprensibles (`validar_stock()`, `calcular_impuestos()`, `cobrar_pasarela()`, `enviar_factura_pdf()`).
3. **Pruebas Automatizadas (*Unit Testing*) y Depuración Rápida**:
   * Es prácticamente imposible hacer tests automáticos con `pytest` sobre un script lineal lleno de `input()` y variables globales. En cambio, una función bien diseñada recibe datos de prueba y permite verificar con exactitud su salida (`assert calcular_iva(100) == 16.0`).
4. **Encapsulamiento y Modelado del Dominio (con Métodos)**:
   * En arquitecturas medianas y grandes, los métodos permiten que cada entidad del negocio (`Usuario`, `Factura`, `ModeloML`, `DataFrame`) lleve consigo las operaciones válidas que pueden hacerse sobre sus propios datos, evitando que cualquier parte del código corrompa su estado interno.
5. **Integración en Todos los Ecosistemas de Python**:
   * **Ciencia de Datos / IA**: Limpieza de datos en Pandas (`df.dropna()`, `df["col"].apply(mi_funcion)`), entrenamiento de modelos en Scikit-Learn (`modelo.fit(X, y)`, `modelo.predict(X_test)`).
   * **Desarrollo Web y APIs (FastAPI / Django / Flask)**: Cada ruta o endpoint de una API es una función decorada que recibe la petición HTTP y devuelve una respuesta JSON.
   * **Automatización y ETL**: Funciones modulares para extraer datos (`extraer_api()`), transformarlos (`limpiar_registros()`) y cargarlos en base de datos (`guardar_sql()`).

### 7.2 Consejos Profesionales de Desarrollo (Clean Code en Python)

Aplicar estas 8 reglas transformará tu código de nivel principiante a nivel profesional:

1. **Principio de Responsabilidad Única (SRP)**:
   * Una función debe hacer **una sola cosa**. Si tu función se llama `descargar_limpiar_y_enviar_reporte()`, divídela en tres funciones separadas. Si una función supera las 25–30 líneas, casi siempre conviene dividirla.
2. **Nombres con Intención Clara (Verbos para Acciones)**:
   * Usa verbos claros para funciones y métodos que realizan acciones: `calcular_descuento()`, `filtrar_usuarios_activos()`, `guardar_archivo()`.
   * Para funciones que devuelven un booleano (`True`/`False`), usa prefijos como `es_`, `tiene_`, `puede_`: `es_mayor_de_edad()`, `tiene_permisos_admin()`.
3. **Usa Anotaciones de Tipo (*Type Hints*) Siempre que Puedas**:
   * Añade los tipos esperados de entrada y salida: `def calcular_imc(peso_kg: float, altura_m: float) -> float:`. No ralentizan Python y permiten que tu editor (VS Code, PyCharm) detecte errores antes de ejecutar el código.
4. **Prefiere Funciones Puras y Evita Efectos Secundarios (*Side Effects*)**:
   * Una **función pura** siempre devuelve el mismo resultado para los mismos argumentos y **no modifica variables externas ni altera silenciosamente las listas/diccionarios que recibe como entrada**. Si necesitas transformar una lista de entrada, devuelve una lista nueva en lugar de mutar la del usuario sin avisar.
5. **Usa Cláusulas de Guarda (*Guard Clauses / Early Return*)**:
   * En lugar de anidar `if` dentro de `if` dentro de `if` (el temido "código flecha"), valida los errores o casos vacíos al principio de la función y retorna o lanza una excepción de inmediato.
6. **Lanza Excepciones Claras en lugar de Retornar "Valores Mágicos"**:
   * Nunca devuelvas `-1`, `None` o `False` mezclado con números reales para indicar que hubo un error, ya que quien llame a tu función olvidará verificarlo. Lanza una excepción descriptiva como `raise ValueError("El salario no puede ser negativo")`.
7. **Limita el Número de Parámetros (Máximo 4 o 5)**:
   * Si una función necesita recibir 8 parámetros distintos, es señal de que hace demasiadas cosas o de que esos parámetros deberían agruparse en un objeto o `@dataclass`.
8. **Cuándo elegir una Función Simple vs. una Clase con Métodos**:
   * **Regla práctica**: Si solo necesitas transformar datos de entrada en datos de salida sin guardar estado entre llamadas, **usa funciones simples en un módulo**. Si necesitas mantener un estado persistente (`self`) que evoluciona a través de múltiples operaciones, **diseña una clase con métodos**.

````python
# Ejemplo 7: Refactorización Profesional (Código Espagueti vs. Código Limpio / Clean Code)

from typing import List, Dict

# ==============================================================================
# ANTES: Función difícil de mantener (Múltiples responsabilidades, sin tipos,
#        anidamiento profundo, muta datos externos y usa valores mágicos)
# ==============================================================================
def proc(d, min_v):
    # ¿Qué es 'd'? ¿Qué es 'min_v'? Nombre poco descriptivo
    res = []
    if d is not None:
        if len(d) > 0:
            for x in d:
                if "activo" in x and x["activo"] == True:
                    if x["ventas"] >= min_v:
                        # Efecto secundario peligroso: modifica el diccionario original
                        x["bono"] = round(x["ventas"] * 0.10, 2)
                        res.append(x)
            return res
        else:
            return -1  # Valor mágico peligroso: mezcla tipos de retorno (list vs int)
    else:
        return -1


# ==============================================================================
# DESPUÉS: Diseño Profesional (Responsabilidad única, Type Hints, Docstring,
#          Guard Clauses, sin mutar los datos originales de entrada)
# ==============================================================================
PORCENTAJE_BONO_ESTANDAR = 0.10

def calcular_bono_vendedor(monto_ventas: float, porcentaje_bono: float = PORCENTAJE_BONO_ESTANDAR) -> float:
    """Calcula el bono monetario correspondiente a un monto de ventas."""
    if monto_ventas < 0:
        raise ValueError("El monto de ventas no puede ser negativo.")
    return round(monto_ventas * porcentaje_bono, 2)


def obtener_vendedores_con_bono(
    vendedores: List[Dict],
    meta_minima_ventas: float
) -> List[Dict]:
    """
    Filtra los vendedores activos que superaron la meta mínima y devuelve
    una nueva lista de registros incluyendo el cálculo de su bono.
    """
    # Guard Clause (Retorno temprano si la lista está vacía)
    if not vendedores:
        return []

    vendedores_premiados = []
    for vendedor in vendedores:
        # Condiciones claras y legibles con retorno temprano (continue)
        es_activo = vendedor.get("activo", False)
        ventas = vendedor.get("ventas", 0.0)

        if not es_activo or ventas < meta_minima_ventas:
            continue

        # Creamos una copia con {**vendedor, ...} para NO mutar el diccionario original
        registro_con_bono = {
            **vendedor,
            "bono": calcular_bono_vendedor(ventas)
        }
        vendedores_premiados.append(registro_con_bono)

    return vendedores_premiados


# --- Prueba del código refactorizado ---
equipo_ventas = [
    {"nombre": "Laura", "activo": True, "ventas": 15000.0},
    {"nombre": "Pedro", "activo": True, "ventas": 4200.0},
    {"nombre": "Sofía", "activo": False, "ventas": 22000.0},
    {"nombre": "Diego", "activo": True, "ventas": 18500.0},
]

premiados = obtener_vendedores_con_bono(equipo_ventas, meta_minima_ventas=10000.0)

print("Vendedores que alcanzaron la meta y su bono calculado:")
for v in premiados:
    print(f"  - {v['nombre']}: Ventas = ${v['ventas']:,.2f} -> Bono = ${v['bono']:,.2f}")

print("\nVerificación de pureza (el diccionario original NO fue alterado):")
print("  Registro original de Laura:", equipo_ventas[0])
````
