## 🟢 Unidad 30. Enfoque Greedy (Algoritmos Voraces) — *[Minutos 05 a 25]*

### 📜 Origen Histórico
- **1202 (Antecedente):** Fibonacci describió en su *Liber Abaci* el método voraz para descomponer fracciones en [fracciones unitarias egipcias](https://en.wikipedia.org/wiki/Greedy_algorithm_for_Egyptian_fractions).
- **1952–1959 (Consolidación en Computación):** Surgen los algoritmos voraces más famosos de la informática en la [Cronología de Algoritmos](https://en.wikipedia.org/wiki/Timeline_of_algorithms): la codificación de **David A. Huffman (1952)**, el árbol de expansión mínima de **Joseph Kruskal ([1956](https://en.wikipedia.org/wiki/Kruskal%27s_algorithm))**, el de **Robert Prim (1957)** y el algoritmo de caminos más cortos de **Edsger W. Dijkstra ([formulado en 1956 y publicado en 1959](https://en.wikipedia.org/wiki/Edsger_W._Dijkstra))**.
- **1971 (Teoría Formal):** **Jack Edmonds** publicó el artículo fundacional [*Matroids and the Greedy Algorithm* (1971)](https://en.wikipedia.org/wiki/Jack_Edmonds), demostrando matemáticamente qué estructura (matroides) debe tener un problema para que el enfoque Greedy garantice siempre la solución óptima global.

### 💡 Concepto Clave en 30 Segundos
Un algoritmo **Greedy** construye la solución paso a paso eligiendo siempre **la opción que parece mejor en este instante (óptimo local)**, sin reconsiderar decisiones pasadas ni explorar ramas futuras.
- **Ventaja:** Es extremadamente rápido, usualmente $O(n \log n)$ porque solo requiere ordenar los datos primero.
- **Trampa:** Solo garantiza el óptimo global si el problema cumple la **propiedad de elección voraz**. Si las monedas son `[1, 3, 4]` y quieres dar cambio de `6`, Greedy elige `4 + 1 + 1` (3 monedas), pero el óptimo real es `3 + 3` (2 monedas, como demuestra el análisis de [matroides y cambio de monedas](https://upcommons.upc.edu/bitstreams/dc5ba090-5311-4093-a70c-a42b3f4be7ee/download)).

```python
# 🐍 UNIDAD 30 EN PYTHON:
# 1) Selección de Actividades (Óptimo garantizado con Greedy ordenando por hora de fin)
# 2) Cambio de Monedas (Cuándo funciona y cuándo falla frente a Programación Dinámica)

def seleccion_actividades_greedy(actividades):
    """
    Recibe tuplas (nombre, inicio, fin).
    Regla Voraz: Elegir siempre la actividad que TERMINE MÁS TEMPRANO y no se solape.
    Complejidad: O(n log n) por el ordenamiento.
    """
    # 1. Ordenar por tiempo de finalización (índice 2)
    ordenadas = sorted(actividades, key=lambda x: x[2])
    seleccionadas = []
    fin_actual = -1

    for nombre, inicio, fin in ordenadas:
        if inicio >= fin_actual:      # Elección voraz local: ¿es compatible?
            seleccionadas.append((nombre, inicio, fin))
            fin_actual = fin
    return seleccionadas

def cambio_monedas_greedy(monedas, monto):
    """Elige siempre la moneda de mayor valor posible."""
    monedas_ord = sorted(monedas, reverse=True)
    resultado = []
    for m in monedas_ord:
        while monto >= m:
            monto -= m
            resultado.append(m)
    return resultado

# --- Prueba en vivo para la clase ---
clases_hoy = [("A", 1, 4), ("B", 3, 5), ("C", 0, 6), ("D", 5, 7), ("E", 8, 9), ("F", 5, 9)]
print("✅ Máximo de actividades compatibles (Greedy):", seleccion_actividades_greedy(clases_hoy))

print("✅ Cambio Greedy en sistema estándar [1, 5, 10, 25] para 63¢:", cambio_monedas_greedy([1, 5, 10, 25], 63))
print("⚠️ Contraejemplo donde Greedy NO es óptimo (monedas [1, 3, 4] para monto 6):",
      cambio_monedas_greedy([1, 3, 4], 6), "-> Usa 3 monedas en vez de [3, 3] (2 monedas).")
```

## 🔵 Unidad 31. Divide y Vencerás (*Divide and Conquer*) — *[Minutos 25 a 45]*

### 📜 Origen Histórico
- **1805 (Precursor matemático):** **Carl Friedrich Gauss** formuló el método que hoy conocemos como Transformada Rápida de Fourier (FFT), redescubierto por Cooley y Tukey en 1965 ([Historia de Divide y Vencerás](https://en.wikipedia.org/wiki/Divide-and-conquer_algorithm)).
- **1945 (Nacimiento en Computación):** El matemático **John von Neumann** inventó y analizó formalmente el algoritmo **Merge Sort** en **1945** para la computadora EDVAC, convirtiéndolo en el primer algoritmo de dos subproblemas diseñado para ordenadores modernos ([Divide-and-conquer algorithm - Wikipedia](https://en.wikipedia.org/wiki/Divide-and-conquer_algorithm)).
- **1960–1969 (Época Dorada):** **Anatolii Karatsuba (1960)** rompió la barrera cuadrática de la multiplicación de números grandes en $O(n^{\log_2 3})$, **C. A. R. Hoare (1962)** creó **Quicksort**, y **Volker Strassen (1969)** aplicó el paradigma a la multiplicación de matrices ([Timeline of Algorithms](https://en.wikipedia.org/wiki/Timeline_of_algorithms)).

### 💡 Concepto Clave en 30 Segundos
Consta de **3 pasos recursivos**:
1. **Dividir (*Divide*):** Partir el problema en subproblemas más pequeños e **independientes** (no se solapan entre sí).
2. **Conquistar (*Conquer*):** Resolver cada subproblema recursivamente hasta llegar a un caso base trivial (ej. una lista de 1 elemento).
3. **Combinar (*Combine*):** Unir las respuestas parciales para formar la solución total.

```python
# 🐍 UNIDAD 31 EN PYTHON:
# 1) Búsqueda Binaria O(log n) — Reducción a 1 subproblema
# 2) Merge Sort O(n log n) de John von Neumann (1945) — División en 2 subproblemas independientes

def busqueda_binaria_dc(lista, objetivo, izq=0, der=None):
    """Busca en O(log n) descartando la mitad del arreglo en cada paso."""
    if der is None:
        der = len(lista) - 1
    if izq > der:
        return -1  # Caso base: no encontrado

    medio = (izq + der) // 2
    if lista[medio] == objetivo:
        return medio
    elif objetivo < lista[medio]:
        return busqueda_binaria_dc(lista, objetivo, izq, medio - 1)  # Conquista mitad izquierda
    else:
        return busqueda_binaria_dc(lista, objetivo, medio + 1, der)  # Conquista mitad derecha

def merge_sort_dc(arr):
    """Ordenamiento Merge Sort O(n log n) mediante Divide y Vencerás."""
    if len(arr) <= 1:
        return arr  # Caso base: 0 o 1 elemento ya está ordenado

    # 1. DIVIDIR en dos mitades independientes
    mitad = len(arr) // 2
    izquierda = merge_sort_dc(arr[:mitad])   # 2. CONQUISTAR izquierda
    derecha = merge_sort_dc(arr[mitad:])     # 2. CONQUISTAR derecha

    # 3. COMBINAR (Merge) ambas mitades ordenadas
    resultado = []
    i = j = 0
    while i < len(izquierda) and j < len(derecha):
        if izquierda[i] <= derecha[j]:
            resultado.append(izquierda[i]); i += 1
        else:
            resultado.append(derecha[j]); j += 1
    return resultado + izquierda[i:] + derecha[j:]

# --- Prueba en vivo para la clase ---
datos = [38, 27, 43, 3, 9, 82, 10]
ordenados = merge_sort_dc(datos)
pos = busqueda_binaria_dc(ordenados, 43)
print("🔹 Lista original :", datos)
print("✅ Merge Sort     :", ordenados)
print(f"✅ Búsqueda Binaria: El número 43 está en el índice {pos} de la lista ordenada.")
```

## 🟠 Unidad 32. Programación Dinámica (*Dynamic Programming*) — *[Minutos 45 a 70]*

### 📜 Origen Histórico
- **1949–1953 (Creación):** Creada por el matemático estadounidense **Richard E. Bellman** mientras trabajaba en procesos de decisión multietapa en la RAND Corporation ([Richard Bellman - Wikipedia](https://en.wikipedia.org/wiki/Richard_Bellman)).
- **1953 y 1957 (Publicaciones Clave):** Bellman introdujo el concepto en su reporte de **1953** (*An Introduction to the Theory of Dynamic Programming* y su artículo *Bottleneck Problems and Dynamic Programming* en *PNAS*) y publicó su obra maestra [*Dynamic Programming* (1957)](https://gwern.net/doc/statistics/decision/1957-bellman-dynamicprogramming.pdf). Eligió la palabra *"dinámica"* para enfatizar sistemas que cambian en el tiempo y *"programación"* en el sentido militar de planificación óptima.

### 💡 Concepto Clave en 30 Segundos
A diferencia de *Divide y Vencerás* (donde las mitades son independientes), usamos **Programación Dinámica** cuando los subproblemas **se solapan (se repiten una y otra vez)** y el problema tiene [subestructura óptima](https://en.wikipedia.org/wiki/Optimal_substructure).  
En vez de recalcular lo mismo millones de veces ($O(2^n)$), **guardamos cada respuesta calculada** de dos formas:
1. **Top-Down (Memoización):** Recursión normal + un diccionario o el decorador `@functools.lru_cache` de Python.
2. **Bottom-Up (Tabulación):** Llenar una lista o tabla iterativamente desde el caso más pequeño `dp[0]` hasta `dp[n]`.

```python
# 🐍 UNIDAD 32 EN PYTHON:
# 1) Fibonacci: Comparación de llamadas entre Recursión Pura O(2^n) vs Memoización O(n)
# 2) El Problema del Cambio de Monedas [1, 3, 4] para monto 6 (Donde Greedy falló en la Unidad 30)

from functools import lru_cache

# Contadores didácticos para mostrar en clase cuántas veces se llama a la función
llamadas_ingenuas = 0
llamadas_memo = 0

def fib_ingenuo(n):
    global llamadas_ingenuas
    llamadas_ingenuas += 1
    if n <= 1:
        return n
    return fib_ingenuo(n - 1) + fib_ingenuo(n - 2)

@lru_cache(maxsize=None)  # <-- ¡En Python basta 1 línea para aplicar Memoización (Top-Down)!
def fib_dinamico_memo(n):
    global llamadas_memo
    llamadas_memo += 1
    if n <= 1:
        return n
    return fib_dinamico_memo(n - 1) + fib_dinamico_memo(n - 2)

def cambio_minimo_tabulacion(monedas, monto):
    """
    Enfoque Bottom-Up (Tabulación): Resuelve de forma exacta el cambio mínimo
    incluso para el sistema [1, 3, 4] y monto 6 donde Greedy fallaba.
    """
    dp = [float('inf')] * (monto + 1)
    dp[0] = 0  # Caso base: 0 monedas para monto 0

    for i in range(1, monto + 1):
        for m in monedas:
            if i - m >= 0:
                dp[i] = min(dp[i], dp[i - m] + 1)
    return dp[monto]

# --- Prueba en vivo para la clase ---
N = 25
res1 = fib_ingenuo(N)
res2 = fib_dinamico_memo(N)
print(f"🔥 Fibonacci({N}) = {res2}")
print(f"   • Llamadas recursivas SIN Programación Dinámica : {llamadas_ingenuas:,} llamadas")
print(f"   • Llamadas recursivas CON Memoización (@cache) : {llamadas_memo} llamadas (¡Ahorro del 99.98%!)")

min_monedas = cambio_minimo_tabulacion([1, 3, 4], 6)
print(f"\n✅ Cambio óptimo con Programación Dinámica para monto 6 con monedas [1, 3, 4]: {min_monedas} monedas (3 + 3).")
```

## 🟣 Unidad 33. Backtracking (Vuelta Atrás) — *[Minutos 70 a 90]*

### 📜 Origen Histórico
- **1848 (El Problema Clásico):** El ajedrecista alemán **Max Bezzel** publicó en **1848** el [problema de las 8 Reinas](https://en.wikipedia.org/wiki/Eight_queens_puzzle) (colocar 8 reinas en un tablero sin que se ataquen), resuelto por Franz Nauck en 1850 y estudiado por Gauss.
- **Década de 1950 (Creación del Término):** El matemático estadounidense **Derrick Henry Lehmer** acuñó formalmente el término **"backtrack"** en los **años 1950** ([Backtracking - Wikipedia](https://en.wikipedia.org/wiki/Backtracking)).
- **1960–1972 (Algoritmos Generales y Lenguajes):** **R. J. Walker (1960)** y **S. W. Golomb con L. D. Baumert (1965)** formalizaron su estructura general de búsqueda en árbol; el lenguaje **SNOBOL (1962)** lo incorporó de forma nativa, y **Edsger W. Dijkstra (1972)** inmortalizó el algoritmo recursivo de las N-Reinas para enseñar programación estructurada ([Eight Queens History](https://en.wikipedia.org/wiki/Eight_queens_puzzle)).

### 💡 Concepto Clave en 30 Segundos
**Backtracking** es una **búsqueda en profundidad (DFS) inteligente con PODA (*pruning*)**.  
Cuando necesitas encontrar combinaciones, permutaciones, resolver un Sudoku o el problema de las $N$-Reinas, el algoritmo sigue la plantilla universal de **3 pasos**:
1. **Elegir (*Choose*):** Agrega un candidato parcial (ej. coloca una reina en una columna segura).
2. **Explorar (*Explore*):** Llama recursivamente para completar el siguiente paso. Si una rama viola una regla, **se poda de inmediato** (no sigue bajando).
3. **Deshacer (*Un-choose / Backtrack*):** Si el camino no llevó a solución (o para buscar más soluciones), quita la pieza colocada y prueba la siguiente alternativa.

```python
# 🐍 UNIDAD 33 EN PYTHON:
# Plantilla Universal de Backtracking aplicada a:
# 1) Problema de las N-Reinas (N=4) con conteo de nodos podados
# 2) Suma de Subconjuntos Exacta (Encontrar qué números suman un objetivo)

def resolver_n_reinas(n=4):
    soluciones = []
    tablero = []  # tablero[fila] = columna donde está la reina
    nodos_visitados = 0

    def es_seguro(fila_actual, col_nueva):
        for fila_ant, col_ant in enumerate(tablero):
            # Misma columna o misma diagonal
            if col_ant == col_nueva or abs(fila_actual - fila_ant) == abs(col_nueva - col_ant):
                return False
        return True

    def backtrack(fila):
        nonlocal nodos_visitados
        nodos_visitados += 1
        if fila == n:
            soluciones.append(tablero[:])  # ¡Solución completa encontrada!
            return

        for col in range(n):
            if es_seguro(fila, col):       # ✂️ PODA: Solo entra si no se atacan
                tablero.append(col)        # 1. ELEGIR (Choose)
                backtrack(fila + 1)        # 2. EXPLORAR (Explore)
                tablero.pop()              # 3. DESHACER / VUELTA ATRÁS (Backtrack)

    backtrack(0)
    return soluciones, nodos_visitados

def suma_subconjuntos_backtrack(numeros, objetivo):
    """Encuentra todas las combinaciones de 'numeros' que suman exactamente 'objetivo'."""
    soluciones = []
    actual = []

    def backtrack(indice, suma_actual):
        if suma_actual == objetivo:
            soluciones.append(actual[:])
            return
        if suma_actual > objetivo or indice == len(numeros):
            return  # ✂️ PODA: Si ya nos pasamos del objetivo, retrocedemos

        # Opción A: Incluir numeros[indice]
        actual.append(numeros[indice])
        backtrack(indice + 1, suma_actual + numeros[indice])
        actual.pop()  # Backtrack (Deshacer)

        # Opción B: No incluir numeros[indice]
        backtrack(indice + 1, suma_actual)

    backtrack(0, 0)
    return soluciones

# --- Prueba en vivo para la clase ---
sols_4reinas, visitados = resolver_n_reinas(4)
print(f"✅ Soluciones para 4-Reinas: {sols_4reinas}")
print(f"✂️ Eficiencia de la Poda: De 4^4 = 256 configuraciones posibles, Backtracking solo visitó {visitados} nodos.")

subconjuntos = suma_subconjuntos_backtrack([2, 4, 6, 8], objetivo=10)
print(f"✅ Subconjuntos de [2, 4, 6, 8] que suman 10: {subconjuntos}")
```

## 🎯 2. Chuleta de Cierre de Clase (Últimos 5 Minutos): ¿Cómo distinguirlos al instante?

1. **Unidad 30 — Greedy (1952–1959 / Edmonds 1971):** *"Ordeno los datos y tomo siempre el mejor pedazo inmediato sin mirar atrás"*. Úsalo cuando puedas demostrar que la decisión local no arruina el futuro (ej. Selección de Actividades, Kruskal, Dijkstra).
2. **Unidad 31 — Divide y Vencerás (Von Neumann 1945):** *"Parto el problema en mitades independientes, resuelvo cada mitad y combino"*. Úsalo para ordenar (`Merge Sort`, `QuickSort`) o buscar en listas ordenadas (`Búsqueda Binaria`).
3. **Unidad 32 — Programación Dinámica (Bellman 1953):** *"Los subproblemas se repiten; guardo cada resultado en caché (`@lru_cache`) o en una tabla `dp[]`"*. Úsalo en optimización (mínimo/máximo/conteo de formas) cuando Greedy falla.
4. **Unidad 33 — Backtracking (Lehmer años 1950):** *"Pruebo un camino (`append`), exploro recursivamente con poda, y si falla deshago el paso (`pop`)"*. Úsalo cuando pidan **todas las soluciones posibles** o satisfacer restricciones estrictas (N-Reinas, Sudoku, laberintos).
