## 📦 Unidad 34. Uso de Módulos de Python — *[Minutos 00 a 15 (15 min)]*

### 💡 ¿Qué es un módulo y por qué usamos `import pandas as pd`?
Desde su lanzamiento en **febrero de 1991** por **Guido van Rossum** ([Historia de Python](https://en.wikipedia.org/wiki/History_of_Python)), Python organiza el código reutilizable en **módulos** (archivos `.py`) y **paquetes** (colecciones de módulos).
1. **Módulos de la Biblioteca Estándar (*Built-in*):** Vienen preinstalados con Python (`math`, `statistics`, `datetime`, `random`). Son útiles para cálculos pequeños, pero procesan listas elemento por elemento.
2. **Librerías Científicas Externas (El Ecosistema PyData):**
   - `import numpy as np` ([NumPy, 2005](https://en.wikipedia.org/wiki/NumPy)): Arreglos numéricos en C.
   - `import pandas as pd` ([Pandas, 2008](https://en.wikipedia.org/wiki/Pandas_(software))): Tablas de datos heterogéneas y estadística descriptiva.
   - `import matplotlib.pyplot as plt` ([Matplotlib, 2003](https://en.wikipedia.org/wiki/Matplotlib)): Gráficos personalizables.

### Sintaxis Esencial de Importación
| Forma de importar | Ejemplo en Python | ¿Cuándo se recomienda? |
| :--- | :--- | :--- |
| **Módulo con alias (Estándar profesional)** | `import pandas as pd` | **Siempre** para `pd`, `np`, `plt` y `sns` por convención universal. |
| **Importación selectiva** | `from datetime import datetime` | Cuando solo necesitas 1 o 2 funciones específicas de un módulo grande. |
| **Módulo completo sin alias** | `import statistics` | Para módulos cortos de la biblioteca estándar de Python. |

```python
# 🐍 UNIDAD 34 EN PYTHON: Módulos Estándar vs. El Poder Vectorizado de Pandas
import math
import statistics
import numpy as np
import pandas as pd

# Lista de ventas diarias en dólares
ventas_lista = [120.5, 150.0, 180.0, 95.5, 210.0, 150.0]

# 1. Con el módulo estándar 'statistics' de Python
media_py = statistics.mean(ventas_lista)
desv_py = statistics.stdev(ventas_lista)

# 2. Con Pandas (Convirtiendo a Serie vectorizada)
ventas_pd = pd.Series(ventas_lista, name="Ventas_USD")

print(f"🔹 Módulo 'statistics' -> Media: ${media_py:.2f} | Desv. Estándar: ${desv_py:.2f}")
print(f"🚀 Con Pandas Series   -> Media: ${ventas_pd.mean():.2f} | Desv. Estándar: ${ventas_pd.std():.2f}")
print("✅ Ventaja de Pandas: Permite aplicar operaciones a toda la columna sin escribir un solo bucle 'for' ( +16% IVA ):")
print((ventas_pd * 1.16).round(2).to_dict())
```

## 🟢 Unidad 35. Serie Pandas (`pd.Series`) para el Tratamiento de Datos — *[Minutos 15 a 35 (20 min)]*

### 💡 ¿Qué es una `Series` en Pandas?
Una [`pd.Series`](https://pandas.pydata.org/docs/reference/api/pandas.Series.describe.html) es una **estructura unidimensional (1D)** que almacena una secuencia de valores junto con un conjunto de etiquetas llamado **índice (`index`)** ([Pandas Data Model](https://en.wikipedia.org/wiki/Pandas_(software))).  
Piensa en ella como **una columna inteligente de Excel** o un diccionario ordenado y vectorizado.

### 📐 Diccionario de Estadística Descriptiva Básica en una `Series`
Para hacer un análisis descriptivo completo sobre una variable numérica o categórica, Pandas ofrece todos los métodos estadísticos listos para usar:

1. **Medidas de Tendencia Central (¿Dónde se concentran los datos?):**
   - **Media aritmética (`.mean()`):** El promedio ($ar{x} = rac{\sum x_i}{n}$). Es sensible a valores extremos (*outliers*).
   - **Mediana (`.median()`):** El valor central ($Q_2$ o percentil 50) cuando los datos están ordenados. **Es robusta frente a valores atípicos**.
   - **Moda (`.mode()`):** El valor o categoría que más se repite.
2. **Medidas de Dispersión (¿Qué tan separados están los datos?):**
   - **Desviación estándar (`.std()`) y Varianza (`.var()`):** Miden cuánto se alejan en promedio los datos respecto a la media.
   - **Rango (`.max() - .min()`):** Distancia entre el valor máximo y el mínimo.
   - **Cuartiles (`.quantile([0.25, 0.5, 0.75])`) y Rango Intercuartílico ($	ext{IQR} = Q_3 - Q_1$):** El $	ext{IQR}$ abarca el 50% central de los datos y sirve para detectar valores atípicos (*outliers*).

```python
# 🐍 UNIDAD 35 EN PYTHON: Creación de Series y Estadística Descriptiva Completa (1D)
# Creamos una Serie con los ingresos mensuales (en miles de USD) de 7 sucursales
# Incluimos una sucursal con ingreso extremo (Sucursal_VIP = 95.0) para comparar Media vs. Mediana:
ingresos = pd.Series(
    data=[22.5, 25.0, 21.0, 25.0, 24.0, 26.5, 95.0],
    index=["Suc_Norte", "Suc_Sur", "Suc_Este", "Suc_Oeste", "Suc_Centro", "Suc_Costa", "Suc_VIP"],
    name="Ingresos_Miles_USD"
)

# Cálculo de Estadísticos Descriptivos Básicos
media = ingresos.mean()
mediana = ingresos.median()
moda = ingresos.mode().iloc[0]
desv_est = ingresos.std()
q1 = ingresos.quantile(0.25)
q3 = ingresos.quantile(0.75)
iqr = q3 - q1

print("📊 1. SERIE CON ÍNDICE ETIQUETADO:")
print(ingresos)

print("\n📐 2. ESTADÍSTICOS DESCRIPTIVOS MANUALES:")
print(f"   • Media (Promedio)      : {media:.2f} (Afectada por la Sucursal VIP de 95.0)")
print(f"   • Mediana (Percentil 50): {mediana:.2f} (Representa mejor a la sucursal típica)")
print(f"   • Moda (Más frecuente)  : {moda:.2f}")
print(f"   • Desviación Estándar   : {desv_est:.2f} | Rango Intercuartílico (IQR = Q3 - Q1): {iqr:.2f}")

print("\n⚡ 3. RESUMEN AUTOMÁTICO CON .describe() EN UNA SOLA LÍNEA:")
print(ingresos.describe().round(2))
```

## 🔵 Unidad 36. Pandas `DataFrame` para Procesamiento de Datos — *[Minutos 35 a 65 (30 min - Núcleo)]*

### 💡 ¿Qué es un `DataFrame`?
Un [`pd.DataFrame`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.html) es una **tabla bidimensional (2D)** compuesta por filas y columnas que comparten un mismo índice. Cada columna es una `Series` y puede tener un tipo de dato distinto (`float64`, `int64`, `string`, `datetime64`, `bool`).

### 🧭 Las 5 Operaciones de Supervivencia en un `DataFrame`
1. **Inspección rápida:** `df.head()`, `df.info()`, `df.shape` (filas, columnas).
2. **Selección sin ambigüedad ([Indexing in Pandas](https://en.wikipedia.org/wiki/Pandas_(software))):**
   - `df.loc[filas, columnas]`: Selecciona usando **nombres/etiquetas** o condiciones booleanas.
   - `df.iloc[filas, columnas]`: Selecciona usando **posiciones enteras** ($0, 1, 2, \dots$).
3. **Limpieza de datos faltantes (`NaN`):** `df.isna().sum()` para contar nulos, `df.fillna(valor)` para imputarlos (ej. con la mediana) o [`df.dropna()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.drop.html) para eliminarlos.
4. **Creación de columnas calculadas:** `df["Margen"] = df["Ventas"] - df["Costo"]`.
5. **Estadística descriptiva global y segmentada:** [`df.describe()`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.describe.html) para toda la tabla y `df.groupby("Categoria").agg(...)` para obtener estadísticos por grupo.

```python
# 🐍 UNIDAD 36 EN PYTHON: Creación, Limpieza de Nulos, Filtrado y Estadística por Grupos en un DataFrame
datos_tienda = {
    "Producto": ["Laptop Pro", "Mouse Óptico", "Silla Ergonómica", "Monitor 4K", "Lámpara LED", "Teclado Mecánico", "Pesa Rusa", "Bicicleta Fija"],
    "Categoria": ["Electrónica", "Electrónica", "Hogar", "Electrónica", "Hogar", "Electrónica", "Deportes", "Deportes"],
    "Precio_USD": [1200.0, 25.0, 280.0, 450.0, None, 110.0, 45.0, 390.0],  # Nota el valor faltante (None -> NaN)
    "Unidades_Vendidas": [15, 120, 30, 40, 85, 60, 50, 22],
    "Calificacion": [4.8, 4.5, 4.6, 4.7, 4.2, 4.9, 4.4, 4.6]
}

df = pd.DataFrame(datos_tienda)

# 1. Limpieza de datos: Rellenamos el precio faltante de 'Lámpara LED' con la mediana de su categoría
mediana_hogar = df["Precio_USD"].median()
df["Precio_USD"] = df["Precio_USD"].fillna(mediana_hogar)

# 2. Columna calculada: Ingreso Total por producto
df["Ingreso_Total"] = df["Precio_USD"] * df["Unidades_Vendidas"]

print("📋 1. VISTA DEL DATAFRAME LIMPIO Y CON COLUMNA CALCULADA:")
print(df.to_string(index=False))

print("\n📐 2. ESTADÍSTICA DESCRIPTIVA MULTIVARIABLE (.describe()):")
print(df[["Precio_USD", "Unidades_Vendidas", "Ingreso_Total", "Calificacion"]].describe().round(2))

print("\n📊 3. ESTADÍSTICA DESCRIPTIVA AGRUPADA POR CATEGORÍA (.groupby()):")
resumen_cat = df.groupby("Categoria").agg(
    Total_Productos=("Producto", "count"),
    Precio_Promedio=("Precio_USD", "mean"),
    Ingreso_Promedio=("Ingreso_Total", "mean"),
    Ingreso_Suma=("Ingreso_Total", "sum"),
    Calificacion_Media=("Calificacion", "mean")
).round(2)
print(resumen_cat)
```

## 🟠 Unidad 37. Ordenación de Datos — *[Minutos 65 a 80 (15 min)]*

### 💡 Cómo Ordenar y Crear Rankings en Pandas
Una vez calculadas las métricas, necesitamos ordenar el `DataFrame` para responder preguntas de negocio como *"¿Cuáles son los 3 productos que generan más ingresos?"* o *"¿Cómo ordeno por Categoría alfabéticamente y dentro de cada categoría de mayor a menor venta?"*:

1. **[`df.sort_values(by=..., ascending=...)`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.sort_values.html):** Ordena por una o varias columnas. Permite algoritmos estables como `kind="mergesort"` y controlar dónde quedan los nulos con `na_position="last"`.
2. **[`df.sort_index()`](https://pandas.pydata.org/docs/reference/api/pandas.Series.sort_index.html):** Ordena las filas o columnas según las etiquetas de su índice (fundamental en series temporales).
3. **[`df.nlargest(n, col)` y `df.nsmallest(n, col)`](https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.nsmallest.html):** Atajos optimizados en C que obtienen el Top $N$ mayor o menor sin tener que ordenar toda la tabla completa.

```python
# 🐍 UNIDAD 37 EN PYTHON: Ordenación Simple, Multicriterio y Top-N Eficiente

# 1. Ordenación Multicriterio: Primero por 'Categoria' (A-Z -> True) y luego por 'Ingreso_Total' (Mayor a menor -> False)
df_ordenado = df.sort_values(by=["Categoria", "Ingreso_Total"], ascending=[True, False])

print("🏆 1. ORDENACIÓN MULTICRITERIO (Categoría A-Z y luego Ingreso de Mayor a Menor):")
print(df_ordenado[["Categoria", "Producto", "Ingreso_Total", "Calificacion"]].to_string(index=False))

# 2. Top 3 productos con mayores ingresos usando .nlargest() y menores usando .nsmallest()
print("\n🥇 2. TOP 3 PRODUCTOS CON MAYOR INGRESO (.nlargest(3)):")
print(df.nlargest(3, "Ingreso_Total")[["Producto", "Categoria", "Ingreso_Total"]].to_string(index=False))

print("\n📉 3. TOP 2 PRODUCTOS MÁS ECONÓMICOS (.nsmallest(2)):")
print(df.nsmallest(2, "Precio_USD")[["Producto", "Precio_USD", "Unidades_Vendidas"]].to_string(index=False))
```

## 🟣 Unidad 38. Datos de Series Temporales *(Electiva)* — *[Minutos 80 a 100 (20 min)]*

### 💡 El ADN Original de Pandas
Recuerda que Wes McKinney creó Pandas en 2008 específicamente para analizar **series financieras en el tiempo** ([Pandas Time-Series Functionality](https://en.wikipedia.org/wiki/Pandas_(software))). Cuando conviertes una columna de texto a fecha con `pd.to_datetime()` y la colocas como índice (`DatetimeIndex`), desbloqueas 4 superpoderes:

1. **Filtrado semántico por fechas:** `df_ts.loc["2026-02"]` devuelve todo febrero automáticamente.
2. **Extracción de atributos temporales (`.dt`):** `df["Fecha"].dt.month`, `.dt.day_name()`, `.dt.quarter`.
3. **Remuestreo temporal (`.resample()`):** Es un `groupby` inteligente para el tiempo. Convierte datos diarios en resúmenes semanales (`"W"`), mensuales (`"ME"`) o trimestrales (`"QE"`).
4. **Ventanas móviles (`.rolling(window=7)`):** Calcula promedios móviles de 7 o 30 días para eliminar el ruido diario y descubrir la tendencia real.

```python
# 🐍 UNIDAD 38 EN PYTHON: Series Temporales, Resampling Mensual y Media Móvil de 7 Días
np.random.seed(42)

# Generamos 90 días de datos (Enero a Marzo de 2026)
fechas_90d = pd.date_range(start="2026-01-01", periods=90, freq="D")
categorias_90d = np.random.choice(["Electrónica", "Hogar", "Deportes", "Libros"], size=90, p=[0.35, 0.30, 0.20, 0.15])
mapa_base = {"Electrónica": 420, "Hogar": 240, "Deportes": 180, "Libros": 65}
ventas_90d = [round(np.random.normal(loc=mapa_base[c], scale=mapa_base[c]*0.22), 2) for c in categorias_90d]
unidades_90d = [max(1, int(round(v / (mapa_base[c]*0.25) + np.random.normal(0, 1)))) for v, c in zip(ventas_90d, categorias_90d)]
satisfaccion_90d = np.clip(np.round(np.random.normal(4.2, 0.45, size=90), 1), 2.5, 5.0)

df_historico = pd.DataFrame({
    "Fecha": fechas_90d,
    "Categoria": categorias_90d,
    "Ventas_USD": ventas_90d,
    "Unidades": unidades_90d,
    "Satisfaccion": satisfaccion_90d
})

# Convertimos 'Fecha' en el índice temporal (DatetimeIndex)
df_ts = df_historico.set_index("Fecha").sort_index()

# 1. Calculamos la Media Móvil de 7 días (.rolling(7).mean())
df_ts["Media_Movil_7D"] = df_ts["Ventas_USD"].rolling(window=7).mean().round(2)

# 2. Remuestreo Mensual (.resample('ME')): Estadísticos descriptivos por cada mes
resumen_mensual = df_ts["Ventas_USD"].resample("ME").agg(
    Dias_Operativos="count",
    Venta_Total_Mes="sum",
    Promedio_Diario="mean",
    Mediana_Diaria="median",
    Desv_Estandar="std"
).round(2)

print("📅 1. RESUMEN ESTADÍSTICO MENSUAL CON .resample('ME'):")
print(resumen_mensual)

print("\n📈 2. PRIMEROS 10 DÍAS CON MEDIA MÓVIL DE 7 DÍAS (.rolling(7).mean()):")
print(df_ts[["Categoria", "Ventas_USD", "Media_Movil_7D"]].head(10))
```
